# Day 13 — DOM & Browser JavaScript

**Goal:** Learn how JavaScript interacts with HTML pages through the DOM.

> **Important:** These cells are JavaScript. Run them in a JavaScript/Jupyter environment, not a Python kernel.

## 1. What is the DOM?

**DOM = Document Object Model**

When the browser loads HTML, it creates an object-based representation of the page.

```text
HTML
  ↓
Browser
  ↓
DOM
  ↓
JavaScript
```

JavaScript uses the DOM to find, inspect, change, create, and remove HTML elements.

## 2. The `document` Object

`document` represents the current webpage.

Try these in the **browser Console**:

```javascript
console.log(document);
console.log(document.title);
```

Change the page title:

```javascript
document.title = "DOM Practice";
```

## 3. Selecting Elements — `getElementById()`

HTML:

```html
<h1 id="title">Welcome</h1>
```

JavaScript:

```javascript
let heading = document.getElementById("title");
console.log(heading);
```

### Important

Correct:

```javascript
document.getElementById("title");
```

Incorrect:

```javascript
document.getElementById("#title");
```

`getElementById()` expects the ID value without `#`.

## 4. `querySelector()`

`querySelector()` returns the **first matching element**.

### Tag selector

```javascript
document.querySelector("h1");
```

### ID selector

```javascript
document.querySelector("#title");
```

### Class selector

```javascript
document.querySelector(".message");
```

### Attribute selector

```javascript
document.querySelector('input[name="email"]');
```

## 5. `querySelectorAll()`

`querySelectorAll()` returns **all matching elements**.

```javascript
let items = document.querySelectorAll(".item");

console.log(items);
console.log(items[0]);
console.log(items[1]);
```

Example:

```html
<p class="item">Apple</p>
<p class="item">Banana</p>
<p class="item">Orange</p>
```

`querySelector()` → first match

`querySelectorAll()` → all matches

## 6. CSS Selectors You Need

| Selector | Meaning | Example |
|---|---|---|
| `button` | Tag | `querySelector("button")` |
| `#login` | ID | `querySelector("#login")` |
| `.username` | Class | `querySelector(".username")` |
| `[name="email"]` | Attribute | `querySelector('[name="email"]')` |
| `.btn.primary` | Multiple classes | `querySelector(".btn.primary")` |
| `.form input` | Descendant | `querySelector(".form input")` |

## 7. Reading Text — `textContent`

HTML:

```html
<h1 id="title">Hello World</h1>
```

JavaScript:

```javascript
let title = document.querySelector("#title");

console.log(title.textContent);
```

Output:

```text
Hello World
```

## 8. Changing Text

```javascript
let title = document.querySelector("#title");

title.textContent = "Welcome Arun";
```

The webpage changes immediately.

```text
Before → Hello
After  → Welcome Arun
```

## 9. `textContent` vs `innerHTML`

### `textContent`

Treats the value as text:

```javascript
element.textContent = "<b>Hello</b>";
```

The page displays:

```text
<b>Hello</b>
```

### `innerHTML`

Treats the value as HTML:

```javascript
element.innerHTML = "<b>Hello</b>";
```

The page displays **Hello** in bold.

```text
textContent → TEXT
innerHTML   → HTML
```

## 10. Changing Styles

```javascript
let title = document.querySelector("#title");

title.style.color = "red";
title.style.backgroundColor = "yellow";
title.style.fontSize = "40px";
```

### CSS vs JavaScript property names

```text
background-color → backgroundColor
font-size        → fontSize
margin-top       → marginTop
border-radius    → borderRadius
```

## 11. Working with Classes

```javascript
let button = document.querySelector("#btn");

button.classList.add("active");

button.classList.remove("active");

button.classList.contains("active");

button.classList.toggle("active");
```

| Method | Purpose |
|---|---|
| `add()` | Add a class |
| `remove()` | Remove a class |
| `contains()` | Check whether class exists |
| `toggle()` | Add if absent, remove if present |

## 12. Working with Attributes

HTML:

```html
<input id="email" type="text" placeholder="Enter email">
```

Read an attribute:

```javascript
let input = document.querySelector("#email");

console.log(input.getAttribute("placeholder"));
console.log(input.getAttribute("type"));
```

Change an attribute:

```javascript
input.setAttribute("placeholder", "Enter your email");
```

Remove an attribute:

```javascript
input.removeAttribute("placeholder");
```

## 13. Input Elements — `.value`

For form controls such as `<input>`, use `.value`.

```javascript
let username = document.querySelector("#username");

console.log(username.value);

username.value = "Arun";

console.log(username.value);
```

### Important

```text
<input> / <textarea> / <select>
            ↓
          value

<h1> / <p> / <div> / <span>
            ↓
       textContent
```

## 14. Creating Elements

Create an element:

```javascript
let paragraph = document.createElement("p");
```

Add text:

```javascript
paragraph.textContent = "Hello from JavaScript";
```

Add it to the page:

```javascript
let container = document.querySelector("#container");

container.appendChild(paragraph);
```

Basic pattern:

```text
create
  ↓
configure
  ↓
append
```

## 15. Removing Elements

```javascript
let message = document.querySelector("#message");

message.remove();
```

The element is removed from the page.

## 16. Clicking Elements

```javascript
let button = document.querySelector("#btn");

button.click();
```

This programmatically triggers a click on the element.

## 17. DOM Events — `addEventListener()`

HTML:

```html
<button id="btn">Click Me</button>
```

JavaScript:

```javascript
let btn = document.querySelector("#btn");

btn.addEventListener("click", () => {
    console.log("Button clicked");
});
```

Common events:

```text
click
input
change
submit
mouseover
keydown
keyup
```

## 18. Button Event Example

HTML:

```html
<button id="btn">Click Me</button>
<p id="message"></p>
```

JavaScript:

```javascript
let btn = document.querySelector("#btn");
let message = document.querySelector("#message");

btn.addEventListener("click", () => {
    message.textContent = "You clicked the button!";
});
```

## 19. Input Event Example

HTML:

```html
<input id="name">
<p id="output"></p>
```

JavaScript:

```javascript
let nameInput = document.querySelector("#name");
let output = document.querySelector("#output");

nameInput.addEventListener("input", () => {
    output.textContent = nameInput.value;
});
```

Flow:

```text
User types
   ↓
input event
   ↓
JavaScript runs
   ↓
DOM changes
```

## 20. DOM Traversal

Useful properties:

```javascript
element.parentElement
element.children
element.firstElementChild
element.lastElementChild
element.nextElementSibling
element.previousElementSibling
```

Example:

```javascript
let container = document.querySelector("#container");

console.log(container.children);
console.log(container.firstElementChild);
console.log(container.lastElementChild);
```

## 21. `querySelectorAll()` + `forEach()`

This combines DOM with the JavaScript you already learned.

```javascript
let items = document.querySelectorAll(".item");

items.forEach(item => {
    console.log(item.textContent);
});
```

You can also modify every matching element:

```javascript
items.forEach(item => {
    item.style.fontWeight = "bold";
});
```

## 22. Complete Login-Style Example

HTML:

```html
<input id="username">
<input id="password">
<button id="login">Login</button>
```

JavaScript:

```javascript
let username = document.querySelector("#username");
let password = document.querySelector("#password");
let login = document.querySelector("#login");

username.value = "arun";
password.value = "12345";

login.click();
```

The automation pattern is:

```text
Find element
     ↓
Enter data
     ↓
Click element
     ↓
Verify result
```

| Change style | `element.style.property = ...` |
| Add class | `element.classList.add()` |
| Remove class | `element.classList.remove()` |
| Check class | `element.classList.contains()` |
| Read attribute | `element.getAttribute()` |
| Change attribute | `element.setAttribute()` |
| Remove attribute | `element.removeAttribute()` |
| Create element | `document.createElement()` |
| Add element | `element.appendChild()` |
| Remove element | `element.remove()` |
| Click | `element.click()` |
| Listen for event | `element.addEventListener()` |
