# Part 2 — `if` Statement

An `if` statement allows JavaScript to make a decision.

### Syntax

```javascript
if (condition) {
    // code
}
```

````typescript
let age = 20;

if (age >= 18) {
    console.log("You are an adult");
}
````

The condition:

```javascript
age >= 18
```

produces `true`, so JavaScript executes the code inside `{ }`.


````typescript
let temperature = 35;

if (temperature > 30) {
    console.log("It is hot");
}
````

````typescript
let marks = 80;

if (marks >= 50) {
    console.log("Pass");
}
````

# Part 3 — `if...else`

`else` gives JavaScript a second path when the condition is false.

### Syntax

```javascript
if (condition) {
    // true
} else {
    // false
}
```

````typescript
let age = 16;

if (age >= 18) {
    console.log("Adult");
} else {
    console.log("Minor");
}
````

````typescript
let marks = 45;

if (marks >= 50) {
    console.log("Pass");
} else {
    console.log("Fail");
}
````

# Part 4 — `else if`

Use `else if` when there are multiple possible conditions.

### Syntax

```javascript
if (condition1) {

} else if (condition2) {

} else {

}
```

````typescript
let marks = 85;

if (marks >= 90) {
    console.log("A");
} else if (marks >= 75) {
    console.log("B");
} else if (marks >= 50) {
    console.log("C");
} else {
    console.log("Fail");
}
````

### Predict the output

```javascript
let marks = 78;
```

The answer is `B`.

### Important

JavaScript checks conditions **from top to bottom**. Once a matching condition is found, that branch runs and the remaining `else if` / `else` branches are skipped.


# Part 5 — Logical Operators in Conditions

This connects directly to Day 1.

## `&&` — AND

Both conditions must be true.


````typescript
let age = 25;

if (age >= 18 && age <= 60) {
    console.log("Working age group");
}
````

## `||` — OR

At least one condition must be true.


````typescript
let day = "Sunday";

if (day === "Saturday" || day === "Sunday") {
    console.log("Weekend");
}
````

## `!` — NOT

Reverses a Boolean value.


````typescript
let isLoggedIn = false;

if (!isLoggedIn) {
    console.log("Please login");
}
````

# Part 6 — Nested `if`

A nested `if` is an `if` statement inside another `if`.


````typescript
let age = 20;
let hasLicense = true;

if (age >= 18) {
    if (hasLicense) {
        console.log("You can drive");
    }
}
````

Sometimes nested conditions can be simplified with logical operators:

````typescript
let age = 20;
let hasLicense = true;

if (age >= 18 && hasLicense) {
    console.log("You can drive");
}
````

# Part 7 — Ternary Operator

The ternary operator is a compact way to write a simple `if...else`.

### Normal version

````typescript
let age = 20;

if (age >= 18) {
    console.log("Adult");
} else {
    console.log("Minor");
}
````

### Ternary version

````typescript
let age = 20;

let result = age >= 18 ? "Adult" : "Minor";

console.log(result);
````

### Syntax

```javascript
condition ? valueIfTrue : valueIfFalse
```

Use the ternary operator mainly for short, simple decisions.

# Part 9 — `for` Loop

The `for` loop is the main loop for Day 2.

### Syntax

```javascript
for (initialization; condition; update) {
    // code
}
```

````typescript
for (let i = 1; i <= 5; i++) {
    console.log(i);
}
````

Output:

```text
1
2
3
4
5
```

### Break it down

1. `let i = 1` → start at 1
2. `i <= 5` → continue while this is true
3. `i++` → increase `i` by 1

Execution:

```text
i = 1 → print
i = 2 → print
i = 3 → print
i = 4 → print
i = 5 → print
i = 6 → stop
```


## `for` Loop Examples

### Print 1–10

````typescript
for (let i = 1; i <= 10; i++) {
    console.log(i);
}
````

### Print even numbers

````typescript
for (let i = 2; i <= 10; i += 2) {
    console.log(i);
}
````

### Countdown

````typescript
for (let i = 5; i >= 1; i--) {
    console.log(i);
}
````

# Part 10 — `while` Loop

A `while` loop repeats while a condition is true.

### Syntax

```javascript
while (condition) {
    // code
}
```

````typescript
let i = 1;

while (i <= 5) {
    console.log(i);
    i++;
}
````

Output:

```text
1
2
3
4
5
```

### Important: avoid infinite loops

The update is important:

```javascript
i++;
```

If you forget it, the condition may remain true forever.

# Final 15-Minute Challenge — Number Analyzer

Given:

```javascript
let number = 15;
```

Write a program that:

1. Checks whether the number is positive, negative, or zero.
2. Checks whether it is even or odd.
3. Prints numbers from `1` to that number.

Expected output:

```text
Positive
Odd
1
2
3
...
15
```


## Possible Solution

Try solving it yourself before running the solution below.


````typescript
let number = 15;

// Positive / Negative / Zero
if (number > 0) {
    console.log("Positive");
} else if (number < 0) {
    console.log("Negative");
} else {
    console.log("Zero");
}

// Even / Odd
if (number % 2 === 0) {
    console.log("Even");
} else {
    console.log("Odd");
}

// Print numbers
for (let i = 1; i <= number; i++) {
    console.log(i);
}
````

# 🧠 Teaching Strategy

For a class of 20 students, use this pattern repeatedly:

**Explain → Predict → Code → Run → Change**

Example:

```javascript
let age = 17;

if (age >= 18) {
    console.log("Adult");
} else {
    console.log("Minor");
}
```

Before running it, ask:

> What will the output be?

Then change `age` to `18` and ask again.

Then change it to `25`.

This encourages students to **think about program execution** instead of only copying code.
