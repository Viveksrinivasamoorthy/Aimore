## 1. Multidimensional Arrays

A multidimensional array contains arrays inside an array. Common uses include matrices, cinema seating, game boards, tables, and grids.


````typescript
let classroom = [
    ["Arun", "Priya", "Rahul"],
    ["Divya", "Kumar", "Meena"],
    ["Vijay", "Anu", "Ravi"]
];

console.log(classroom[0][0]);
console.log(classroom[1][2]);
console.log(classroom[2][1]);

for (let i = 0; i < classroom.length; i++) {
    for (let j = 0; j < classroom[i].length; j++) {
        console.log(classroom[i][j]);
    }
}
````

### Use Case — Cinema Seating

````typescript
let seats = [
    ["A1", "A2", "A3"],
    ["B1", "B2", "B3"],
    ["C1", "C2", "C3"]
];

console.log("Selected seat:", seats[1][2]);
````

## 2. Nested Objects

Objects can contain other objects. This structure is common in API responses and application data.


````typescript
let customer = {
    name: "Priya",
    contact: {
        email: "priya@example.com",
        phone: "9876543210"
    },
    address: {
        city: "Chennai",
        state: "Tamil Nadu",
        country: "India"
    }
};

console.log(customer.contact.email);
console.log(customer.address.city);
console.log(customer.address.state);
````

## 3. Objects Containing Arrays

````typescript
let student = {
    name: "Arun",
    subjects: ["JavaScript", "HTML", "CSS", "SQL"],
    marks: [85, 90, 78, 88]
};

console.log(student.subjects[0]);
console.log(student.marks[0]);
````

## 5. Set — Unique Values

`Set` stores unique values. Useful for removing duplicates, unique IDs, tags, usernames, and email addresses.


````typescript
let numbers = [10, 20, 10, 30, 20, 40,50,60,60];

// let uniqueNumbers = new Set(numbers);
let uniqueNumbers = [...new Set(numbers)];

console.log(uniqueNumbers);
// console.log([...uniqueNumbers]);
````

### Use Case — Remove Duplicate Emails

````typescript
let emails = [
    "a@gmail.com", "b@gmail.com", "a@gmail.com",
    "c@gmail.com", "b@gmail.com"
];

let uniqueEmails = [...new Set(emails)];

console.log(uniqueEmails);
````

## 6. Map — Key-Value Lookup

````typescript
let students = new Map();

students.set(101, "Arun");
students.set(102, "Priya");
students.set(103, "Rahul");
console.log(students);

console.log("Student 101:", students.get(101));
console.log("Has 102:", students.has(102));

students.delete(103);

console.log(students);
````

### Use Case — Employee ID Lookup

````typescript
let employees = new Map();

employees.set(1001, "Arun");
employees.set(1002, "Priya");
employees.set(1003, "Rahul");

let employeeId = 1002;

if (employees.has(employeeId)) {
    console.log("Employee:", employees.get(employeeId));
} else {
    console.log("Employee not found");
}
````

## 7. Frequency Counter

A frequency counter counts how often each value occurs. This pattern is useful for word counts, voting, sales analysis, and survey data.


````typescript
let numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4];

let frequency = {};

for (let i = 0; i < numbers.length; i++) {
    let number = numbers[i];

    if (frequency[number] === undefined) {
        frequency[number] = 1;
    } else {
        frequency[number]++;
    }
}

console.log(frequency);
````

### Frequency Counter — Words

````typescript
let words = ["apple", "banana", "apple", "orange", "banana", "apple"];

let wordFrequency = {};

for (let i = 0; i < words.length; i++) {
    let word = words[i];

    if (wordFrequency[word] === undefined) {
        wordFrequency[word] = 1;
    } else {
        wordFrequency[word]++;
    }
}

console.log(wordFrequency);
````

## 8. Stack — LIFO

**Last In, First Out.** Think of a stack of plates. Arrays can implement a stack with `push()` and `pop()`.

**Use cases:** browser history, undo, function call stack, expression evaluation, backtracking.


````typescript
let stack = [];

stack.push("Page 1");
stack.push("Page 2");
stack.push("Page 3");

console.log("Stack:", stack);
console.log("Removed:", stack.pop());
console.log("After pop:", stack);
````

### Use Case — Browser History

````typescript
let history = [];

history.push("google.com");
history.push("youtube.com");
history.push("github.com");

console.log("Current:", history[history.length - 1]);

history.pop();

console.log("After Back:", history[history.length - 1]);
````

## 9. Queue — FIFO

**First In, First Out.** Think of people waiting in a line. Arrays can implement a simple queue with `push()` and `shift()`.

**Use cases:** printer queue, customer service, ticket booking, task processing.


````typescript
let queue = [];

queue.push("Arun");
queue.push("Priya");
queue.push("Rahul");

console.log("Queue:", queue);
console.log("Serving:", queue.shift());
console.log("After serving:", queue);
````

### Use Case — Customer Service Queue

````typescript
let customers = ["Customer 1", "Customer 2", "Customer 3"];

while (customers.length > 0) {
    let customer = customers.shift();
    console.log("Serving:", customer);
}
````

## 10. Sorting Numbers

For numeric sorting, provide a comparison function.


````typescript
let marks = [78, 92, 65, 88, 72];

marks.sort((a, b) => a - b);
console.log("Ascending:", marks);

marks.sort((a, b) => b - a);
console.log("Descending:", marks);

console.log("Default sort:", [10, 2, 30].sort());
console.log("Numeric sort:", [10, 2, 30].sort((a, b) => a - b));
````

## 11. Sorting an Array of Objects

````typescript
let studentsByMark = [
    { name: "Arun", mark: 75 },
    { name: "Priya", mark: 92 },
    { name: "Rahul", mark: 85 }
];

studentsByMark.sort((a, b) => b.mark - a.mark);

console.log(studentsByMark);
````

## 12. Searching with `find()`

`find()` returns the first element that matches a condition.


````typescript
let productsForSearch = [
    { id: 101, name: "Laptop" },
    { id: 102, name: "Mouse" },
    { id: 103, name: "Keyboard" }
];

let product = productsForSearch.find(product => product.id === 102);

console.log(product);
````

## 13. Filtering with `filter()`

`filter()` returns all elements that satisfy a condition.


````typescript
let studentsForFilter = [
    { name: "Arun", mark: 85 },
    { name: "Priya", mark: 92 },
    { name: "Rahul", mark: 35 },
    { name: "Divya", mark: 78 }
];

let passedStudents = studentsForFilter.filter(student => student.mark >= 40);

console.log(passedStudents);
````

### Use Case — Products Under a Budget

````typescript
let productsForBudget = [
    { name: "Laptop", price: 55000 },
    { name: "Mouse", price: 800 },
    { name: "Keyboard", price: 1500 },
    { name: "Monitor", price: 12000 }
];

let affordableProducts = productsForBudget.filter(product => product.price <= 1500);

console.log(affordableProducts);
````

## 14. Transforming Data with `map()`

`map()` creates a new array by transforming every element.


````typescript
let studentsForMap = [
    { name: "Arun", mark: 85 },
    { name: "Priya", mark: 92 },
    { name: "Rahul", mark: 76 }
];

let names = studentsForMap.map(student => student.name);

console.log(names);
````

### Use Case — Apply a Discount

````typescript
let prices = [100, 200, 300];

let discountedPrices = prices.map(price => price * 0.9);

console.log(discountedPrices);
````

## 15. `reduce()` — Aggregate Data

`reduce()` processes an array into one final value. Common uses include totals, sums, counts, maximums, and grouping.


````typescript
let pricesForTotal = [100, 200, 300, 400];

let total = pricesForTotal.reduce(
    (sum, price) => sum + price,
    0
);

console.log("Total:", total);
````

### Use Case — Shopping Cart Total

````typescript
let cart = [
    { name: "Laptop", price: 50000, quantity: 1 },
    { name: "Mouse", price: 800, quantity: 2 },
    { name: "Keyboard", price: 1500, quantity: 1 }
];

let cartTotal = cart.reduce(
    (sum, item) => sum + item.price * item.quantity,
    0
);

console.log("Cart Total:", cartTotal);
````

## 16. Find Highest Value with `reduce()`

````typescript
let marksForMax = [78, 92, 65, 88, 72];

let highestMark = marksForMax.reduce(
    (highest, mark) => mark > highest ? mark : highest,
    marksForMax[0]
);

console.log("Highest:", highestMark);
````

## 17. Method Chaining

A common data-processing pattern is:

`filter()` → `map()` → `reduce()`


````typescript
let productsForChain = [
    { name: "Mouse", price: 800 },
    { name: "Keyboard", price: 1500 },
    { name: "Monitor", price: 12000 },
    { name: "USB Cable", price: 500 }
];

let totalExpensiveProducts = productsForChain
    .filter(product => product.price > 1000)
    .map(product => product.price)
    .reduce((sum, price) => sum + price, 0);

console.log("Total:", totalExpensiveProducts);
````

## 18. API-Style Data Processing

Modern applications commonly receive nested arrays and objects.


````typescript
let users = [
    { id: 1, name: "Arun", age: 25 },
    { id: 2, name: "Priya", age: 22 },
    { id: 3, name: "Rahul", age: 30 }
];

let userNames = users.map(user => user.name);
let adults = users.filter(user => user.age >= 18);
let selectedUser = users.find(user => user.id === 2);

console.log("Names:", userNames);
console.log("Adults:", adults);
console.log("Selected:", selectedUser);
````
