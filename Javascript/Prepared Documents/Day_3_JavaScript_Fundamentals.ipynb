## 2. `if` Statement

If the condition is `true`, JavaScript executes the code inside the block.

````javascript
let age = 20;

if (age >= 18) {
    console.log("You are eligible to vote");
}
````

````javascript
let temperature = 35;

if (temperature > 30) {
    console.log("It is hot");
}
````

## 3. `if...else`

````javascript
let age = 16;

if (age >= 18) {
    console.log("Adult");
} else {
    console.log("Minor");
}
````

````javascript
let number = 7;

if (number % 2 === 0) {
    console.log("Even");
} else {
    console.log("Odd");
}
````

## 4. `else if`

JavaScript checks the conditions from top to bottom and runs the first matching block.

````javascript
let marks = 75;

if (marks >= 90) {
    console.log("A+");
} else if (marks >= 80) {
    console.log("A");
} else if (marks >= 70) {
    console.log("B");
} else if (marks >= 60) {
    console.log("C");
} else {
    console.log("Fail");
}
````

## 5. Nested `if`

````javascript
let age = 25;
let hasLicense = true;

if (age >= 18) {
    if (hasLicense === true) {
        console.log("You can drive");
    }
}
````

### Same logic using `&&`

````javascript
if (age >= 18 && hasLicense === true) {
    console.log("You can drive");
}
````

## 6. Truthy and Falsy Values

Common falsy values:

- `false`
- `0`
- `""`
- `null`
- `undefined`
- `NaN`

````javascript
let username = "";

if (username) {
    console.log("Username entered");
} else {
    console.log("Username is empty");
}
````

## 7. Ternary Operator

Syntax:

```javascript
condition ? valueIfTrue : valueIfFalse
```

````javascript
let age = 20;

let result = age >= 18 ? "Adult" : "Minor";

console.log(result);
````

Use ternary for simple decisions. Avoid forcing complicated logic into a ternary expression.

## 8. `switch` Statement

`switch` is useful when comparing one value against several fixed choices.

````javascript
let day = 2;

switch (day) {
    case 1:
        console.log("Monday");
        break;

    case 2:
        console.log("Tuesday");
        break;

    case 3:
        console.log("Wednesday");
        break;

    default:
        console.log("Invalid day");
}
````

### Why `break`?

Without `break`, JavaScript can continue executing the following cases after a match. `default` runs when no case matches.

# 9. Practical Examples

### Example 1 — Positive, Negative or Zero

````javascript
let number = -5;

if (number > 0) {
    console.log("Positive");
} else if (number < 0) {
    console.log("Negative");
} else {
    console.log("Zero");
}
````

````javascript
let a = 25;
let b = 40;

if (a > b) {
    console.log("a is larger");
} else if (b > a) {
    console.log("b is larger");
} else {
    console.log("Both are equal");
}
````

### Example 3 — Login Check

````javascript
let username = "admin";
let password = "1234";

if (username === "admin" && password === "1234") {
    console.log("Login successful");
} else {
    console.log("Invalid username or password");
}
````

# 11. Mini Project — ATM Simulator

Start with a basic withdrawal program and then extend it with validation.

````javascript
let balance = 5000;
let withdrawal = 2000;

if (withdrawal <= balance) {
    balance = balance - withdrawal;
    console.log("Withdrawal successful");
    console.log("Remaining balance:", balance);
} else {
    console.log("Insufficient balance");
}
````

### Improved ATM validation

````javascript
let balance = 5000;
let withdrawal = 2000;

if (withdrawal <= 0) {
    console.log("Invalid amount");
} else if (withdrawal > balance) {
    console.log("Insufficient balance");
} else {
    balance = balance - withdrawal;
    console.log("Withdrawal successful");
    console.log("Remaining balance:", balance);
}
````
