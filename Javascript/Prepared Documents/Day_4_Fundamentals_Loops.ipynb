# 1. Why Do We Need Loops?

A loop repeats a block of code automatically.

````typescript
for (let i = 1; i <= 5; i++) {
    console.log(i);
}
````

# 2. `for` Loop

```javascript
for (initialization; condition; update) {
    // code to repeat
}
```

````typescript
for (let i = 1; i <= 10; i++) {
    console.log(i);
}
````

### Loop flow

1. Initialize
2. Check condition
3. Execute body
4. Update
5. Repeat until false

# 3. Print Even Numbers

````typescript
for (let i = 1; i <= 20; i++) {
    if (i % 2 === 0) {
        console.log(i);
    }
}
````

# 4. Print Odd Numbers

````typescript
for (let i = 1; i <= 20; i++) {
    if (i % 2 !== 0) {
        console.log(i);
    }
}
````

# 5. Multiplication Table

````typescript
let number = 5;

for (let i = 1; i <= 10; i++) {
    console.log(number + " x " + i + " = " + (number * i));
}
````

# 6. Sum of Numbers — Accumulator Pattern

````typescript
let sum = 0;

for (let i = 1; i <= 10; i++) {
    console.log(sum,i)
    sum = sum + i;
    console.log(sum)
}

console.log("Sum:", sum);
````

# 7. Count Values Matching a Condition

Count how many numbers from 1–100 are divisible by 5.

````typescript
let count = 0;

for (let i = 1; i <= 100; i++) {
    if (i % 5 === 0) {
        count++;
    }
}

console.log("Count:", count);
````

# 8. Factorial

````typescript
let number = 5;
let factorial = 1;

for (let i = 1; i <= number; i++) {
    factorial = factorial * i;
}

console.log("Factorial:", factorial);
````

# 9. Countdown

````typescript
for (let i = 10; i >= 1; i--) {
    console.log(i);
}

console.log("Go!");
````

# 10. `while` Loop

Use `while` when repetition depends mainly on a condition.

````typescript
let i = 1;

while (i <= 5) {
    console.log(i);
    i++;
}
````

## Important

Update the value used by the condition, otherwise you may create an infinite loop.

# 11. `while` Countdown

````typescript
let count = 10;

while (count > 0) {
    console.log(count);
    count--;
}

console.log("Time's up!");
````

# 12. `while` — Password Attempts

````typescript
let attempts = 0;
let correctPassword = "1234";
let enteredPassword = "1111";

while (attempts < 3) {
    if (enteredPassword === correctPassword) {
        console.log("Login successful");
        break;
    }

    attempts++;
    console.log("Incorrect password");
}

if (attempts === 3) {
    console.log("Account locked");
}
````

# 13. `while` — ATM Balance

````typescript
let balance = 5000;
let withdrawal = 1000;

while (balance >= withdrawal) {
    balance = balance - withdrawal;
    console.log("Withdrawal successful");
    console.log("Balance:", balance);
}
````

# 14. `while` — Game Lives

````typescript
let lives = 3;

while (lives > 0) {
    console.log("Game is running...");
    lives--;
}

console.log("Game Over");
````

# 15. `while` — Find the First Number Divisible by 7

````typescript
let number = 1;

while (number % 7 !== 0) {
    number++;
}

console.log("First number divisible by 7:", number);
````

# 16. `break`

`break` immediately exits a loop.

````typescript
for (let i = 1; i <= 10; i++) {
    if (i === 5) {
        break;
    }
    console.log(i);
}
````

### Search example

````typescript
let target = 7;

for (let i = 1; i <= 100; i++) {
    if (i === target) {
        console.log("Found:", i);
        break;
    }
}
````

# 17. `continue`

`continue` skips the current iteration.

````typescript
for (let i = 1; i <= 10; i++) {
    if (i === 5) {
        continue;
    }
    console.log(i);
}
````

# 18. Nested Loops

````typescript
for (let row = 1; row <= 3; row++) {
    for (let column = 1; column <= 3; column++) {
        console.log("Row:", row, "Column:", column);
    }
}
````

# 19. Pattern Printing

````typescript
for (let i = 1; i <= 5; i++) {
    let stars = "";

    for (let j = 1; j <= i; j++) {
        stars = stars + "*";
    }

    console.log(stars);
}
````

# 20. `for` vs `while`

| `for` | `while` |
|---|---|
| Usually useful when repetitions are known | Usually useful when repetition depends on a condition |
| Great for ranges and fixed repetitions | Great for retries and process-until conditions |

Example `for`:
```javascript
for (let i = 1; i <= 10; i++) {
    console.log(i);
}
```

Example `while`:
```javascript
while (balance > 0) {
    // process something
}
```

# 21. Preview — Loops + Arrays

````typescript
let prices = [100, 250, 150];
let total = 0;

for (let i = 0; i < prices.length; i++) {
    total = total + prices[i];
}

console.log("Total:", total);
````

# 22. Student Marks — Total and Average

````typescript
let marks = [75, 82, 90, 68, 88];
let total = 0;

for (let i = 0; i < marks.length; i++) {
    total = total + marks[i];
}

let average = total / marks.length;

console.log("Total:", total);
console.log("Average:", average);
````

# 23. Find Failed Students

````typescript
let marks = [75, 42, 90, 35, 88];

for (let i = 0; i < marks.length; i++) {
    if (marks[i] < 40) {
        console.log("Failed mark:", marks[i]);
    }
}
````

# Day 4 Summary

### Three important loop patterns

**Counting**
```javascript
let count = 0;
for (...) {
    count++;
}
```

**Accumulating**
```javascript
let total = 0;
for (...) {
    total = total + value;
}
```

**Searching**
```javascript
for (...) {
    if (condition) {
        console.log("Found");
        break;
    }
}
```
