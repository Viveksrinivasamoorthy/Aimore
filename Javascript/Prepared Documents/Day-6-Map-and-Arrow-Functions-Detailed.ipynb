# JavaScript Map() and Arrow Functions — Detailed Class

Covers Map APIs, real-world use cases, arrow functions, callbacks, and array methods.

> Use a JavaScript-capable Jupyter kernel such as Deno or Node.js in VS Code.


## Map() Quick Reference

`set()` add/update • `get()` retrieve • `has()` check • `delete()` remove • `clear()` remove all • `size` count • `keys()` keys • `values()` values • `entries()` pairs • `forEach()` iterate.


````javascript
let students = new Map();
students.set(101, "Arun");
students.set(102, "Priya");
students.set(103, "Rahul");
console.log(students);
console.log("get:", students.get(101));
console.log("has:", students.has(102));
console.log("size:", students.size);
console.log("keys:", [...students.keys()]);
console.log("values:", [...students.values()]);
console.log("entries:", [...students.entries()]);
````

## Map Methods in Detail

### `set(key, value)`
Adds a new entry or updates an existing key. Returns the Map, so calls can be chained.

### `get(key)`
Returns the value for a key, or `undefined` when the key is absent.

### `has(key)`
Returns `true` or `false` depending on whether the key exists.

### `delete(key)`
Removes one entry and returns `true` if it existed, otherwise `false`.

### `clear()`
Removes every entry.

### `size`
Number of key-value entries. Unlike arrays, Map uses `size`, not `length`.

### `keys()`
Iterator of keys.

### `values()`
Iterator of values.

### `entries()`
Iterator of `[key, value]` pairs.

### `forEach(callback)`
Iterates through entries; callback arguments are `(value, key, map)`.


````javascript
let m = new Map([[1, "One"], [2, "Two"], [3, "Three"]]);
console.log(m.get(2));
console.log(m.has(3));
console.log(m.delete(1));
console.log(m.size);
m.forEach((value, key) => console.log(key, value));
m.clear();
console.log(m.size);
````

## Map Keys Can Be Any JavaScript Value


````javascript
let data = new Map();
data.set(101, "number key");
data.set("101", "string key");
data.set(true, "boolean key");
let user = {name: "Arun"};
data.set(user, "Logged In");
console.log(data.get(101));
console.log(data.get("101"));
console.log(data.get(user));
````

## Real-World Map Use Cases


````javascript
let employees = new Map([
  ["EMP001", {name:"Arun", department:"IT", salary:50000}],
  ["EMP002", {name:"Priya", department:"HR", salary:45000}],
  ["EMP003", {name:"Rahul", department:"Finance", salary:60000}]
]);
let id = "EMP002";
if (employees.has(id)) console.log(employees.get(id));
````

````javascript
let prices = new Map([["LAPTOP",55000],["MOUSE",800],["KEYBOARD",1500]]);
let code = "MOUSE";
console.log("Price:", prices.get(code));
````

## Frequency Counter with Map


````javascript
let numbers = [10,20,10,30,20,10];
let frequency = new Map();
for (let number of numbers) {
  frequency.set(number, (frequency.get(number) ?? 0) + 1);
}
console.log(frequency);
````

# Arrow Functions

An arrow function is a shorter syntax for a function. The `=>` symbol is the arrow.


````javascript
function addTraditional(a, b) { return a + b; }
const addArrow = (a, b) => { return a + b; };
const square = number => number * number;
console.log(addTraditional(10,20));
console.log(addArrow(10,20));
console.log(square(5));
````

## Explicit vs Implicit Return

Explicit: `const add = (a,b) => { return a+b; };`

Implicit: `const add = (a,b) => a+b;`


````javascript
const calculate = (a,b) => { let total=a+b; console.log(total); return total; };
const multiply = (a,b) => a*b;
console.log(calculate(10,20));
console.log(multiply(5,6));
````

## Arrow Functions as Callback Functions

A function passed to another function is a callback. Arrow functions are commonly used for callbacks.


````javascript
let nums = [1,2,3,4,5];
nums.forEach(number => console.log(number));
let doubled = nums.map(number => number * 2);
let large = nums.filter(number => number > 3);
let found = nums.find(number => number === 4);
let total = nums.reduce((sum, number) => sum + number, 0);
console.log(doubled, large, found, total);
````

## Important: `Map` vs `map()`

`Map` is a **data structure**. `map()` is an **array method** used to transform elements.


````javascript
let studentMap = new Map([[101,"Arun"],[102,"Priya"]]);
console.log(studentMap.get(101));
let names = ["Arun","Priya","Rahul"].map(name => name.toUpperCase());
console.log(names);
````
