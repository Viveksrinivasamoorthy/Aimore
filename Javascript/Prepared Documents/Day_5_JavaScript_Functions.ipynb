## 1. What is a Function?

A function is a reusable block of code designed to perform a specific task.

### Without a function

````javascript
console.log("Welcome to JavaScript");
console.log("Welcome to JavaScript");
console.log("Welcome to JavaScript");
````

### With a function

````javascript
function welcome() {
    console.log("Welcome to JavaScript");
}

welcome();
welcome();
welcome();
````

### Function syntax

```javascript
function functionName() {
    // code
}
```

A function is not executed just because we define it. We execute it by **calling/invoking** it.

## 2. Function Declaration and Function Call

````javascript
function greet() {
    console.log("Hello!");
}

greet();
````

### Terminology

```javascript
function greet() {
    console.log("Hello!");
}
```

- `function` → keyword
- `greet` → function name
- `()` → parameter list
- `{}` → function body

```javascript
greet();
```

- `greet()` → function call / invocation

## 3. Functions with Parameters

````javascript
function greet(name) {
    console.log("Hello " + name);
}

greet("Arun");
greet("Priya");
greet("Rahul");
````

### Parameter vs Argument

```javascript
function greet(name) {
    // name is a parameter
}

greet("Arun");
// "Arun" is an argument
```

## 4. Multiple Parameters

````javascript
function add(a, b) {
    console.log(a + b);
}

add(10, 20);
add(50, 30);
````

````javascript
function introduce(name, age, city) {
    console.log("Name:", name);
    console.log("Age:", age);
    console.log("City:", city);
}

introduce("Arun", 25, "Chennai");
````

## 5. Return Values ⭐

````javascript
function add(a, b) {
    return a + b;
}

let result = add(10, 20);

console.log(result);
````

### Why use `return`?

`return` sends a value back to the code that called the function. That value can then be stored, calculated with, compared, or passed to another function.

````javascript
function add(a, b) {
    return a + b;
}

let total = add(10, 20);
let finalAmount = total + 100;

console.log("Total:", total);
console.log("Final amount:", finalAmount);
````

## 6. `console.log()` vs `return`

````javascript
function addUsingLog(a, b) {
    console.log(a + b);
}

function addUsingReturn(a, b) {
    return a + b;
}

let result1 = addUsingLog(10, 20);
let result2 = addUsingReturn(10, 20);

console.log("result1:", result1);
console.log("result2:", result2);
````

### Important

- `console.log()` displays a value.
- `return` gives a value back to the caller.
- A function without an explicit `return` returns `undefined`.

## 7. Useful Function Examples

````javascript
function square(number) {
    return number * number;
}

console.log(square(5));
console.log(square(10));
````

````javascript
function isEven(number) {
    return number % 2 === 0;
}

console.log(isEven(10));
console.log(isEven(7));
````

````javascript
function maximum(a, b) {
    if (a > b) {
        return a;
    }

    return b;
}

console.log(maximum(10, 20));
console.log(maximum(100, 50));
````

## 8. Functions + Conditions

````javascript
function getGrade(mark) {
    if (mark >= 90) {
        return "A";
    } else if (mark >= 75) {
        return "B";
    } else if (mark >= 60) {
        return "C";
    } else if (mark >= 40) {
        return "D";
    } else {
        return "F";
    }
}

console.log(getGrade(95));
console.log(getGrade(72));
console.log(getGrade(35));
````

## 9. Default Parameters

````javascript
function greet(name = "Guest") {
    console.log("Hello " + name);
}

greet("Arun");
greet();
````

If an argument is not supplied, the default parameter value is used.

## 10. Scope — Global vs Local Variables

````javascript
let company = "ABC";

function employee() {
    let name = "Arun";

    console.log("Inside function - company:", company);
    console.log("Inside function - name:", name);
}

employee();

console.log("Outside function - company:", company);
````

The variable `name` exists only inside `employee()`. Trying to use it outside that function would cause an error.

### Simple rule

- **Global variable** → available from a wider scope
- **Local variable** → available only inside its scope

Prefer keeping variables in the smallest scope where they are needed.

## 11. Function Expression

````javascript
const add = function(a, b) {
    return a + b;
};

console.log(add(10, 20));
````

Here, the function is assigned to the variable `add`.

A function expression is useful when functions need to be stored in variables or passed around as values.

## 12. Arrow Functions ⭐

````javascript
function addTraditional(a, b) {
    return a + b;
}

const addArrow = (a, b) => {
    return a + b;
};

console.log(addTraditional(10, 20));
console.log(addArrow(10, 20));
````

### Short arrow function

````javascript
const add = (a, b) => a + b;

const square = n => n * n;

const isEven = n => n % 2 === 0;

console.log(add(10, 20));
console.log(square(5));
console.log(isEven(10));
````

When an arrow function contains a single expression, the expression can be returned implicitly.

```javascript
const square = n => n * n;
```

is equivalent to:

```javascript
const square = n => {
    return n * n;
};
```

## 13. Functions + Arrays

````javascript
function printArray(numbers) {
    for (let i = 0; i < numbers.length; i++) {
        console.log(numbers[i]);
    }
}

let numbers = [10, 20, 30, 40, 50];

printArray(numbers);
````

### Calculate Array Sum

````javascript
function calculateSum(numbers) {
    let sum = 0;

    for (let i = 0; i < numbers.length; i++) {
        sum += numbers[i];
    }

    return sum;
}

let numbers = [10, 20, 30, 40];

console.log("Sum:", calculateSum(numbers));
````

### Find the Largest Number

````javascript
function findLargest(numbers) {
    let largest = numbers[0];

    for (let i = 1; i < numbers.length; i++) {
        if (numbers[i] > largest) {
            largest = numbers[i];
        }
    }

    return largest;
}

console.log(findLargest([10, 50, 20, 80, 30]));
````

### Count Even Numbers

````javascript
function countEvenNumbers(numbers) {
    let count = 0;

    for (let number of numbers) {
        if (number % 2 === 0) {
            count++;
        }
    }

    return count;
}

console.log(
    countEvenNumbers([10, 15, 20, 23, 30, 35])
);
````

## 14. Functions + Objects

````javascript
let student = {
    name: "Arun",
    age: 21,
    mark: 85
};

function printStudent(student) {
    console.log("Name:", student.name);
    console.log("Age:", student.age);
    console.log("Mark:", student.mark);
}

printStudent(student);
````

## 15. Real-World Use Case — Shopping Cart

````javascript
let cart = [
    {
        name: "Laptop",
        price: 50000
    },
    {
        name: "Mouse",
        price: 1000
    },
    {
        name: "Keyboard",
        price: 2000
    }
];

function calculateCartTotal(cart) {
    let total = 0;

    for (let item of cart) {
        total += item.price;
    }

    return total;
}

console.log("Cart total:", calculateCartTotal(cart));
````

### Add GST

````javascript
function calculateGST(amount) {
    return amount * 0.18;
}

let total = calculateCartTotal(cart);
let gst = calculateGST(total);
let finalAmount = total + gst;

console.log("Total:", total);
console.log("GST:", gst);
console.log("Final amount:", finalAmount);
````

## 16. Real-World Use Case — Login

````javascript
function login(username, password) {
    if (username === "admin" && password === "1234") {
        return "Login successful";
    }

    return "Invalid username or password";
}

console.log(login("admin", "1234"));
console.log(login("user", "1234"));
````

## 17. Real-World Use Case — ATM

````javascript
function checkBalance(balance) {
    return balance;
}

function deposit(balance, amount) {
    return balance + amount;
}

function withdraw(balance, amount) {
    if (amount > balance) {
        return "Insufficient balance";
    }

    return balance - amount;
}

let balance = 10000;

console.log("Balance:", checkBalance(balance));

balance = deposit(balance, 5000);
console.log("After deposit:", balance);

balance = withdraw(balance, 3000);
console.log("After withdrawal:", balance);
````

## 18. Function Composition

````javascript
function square(number) {
    return number * number;
}

function double(number) {
    return number * 2;
}

function process(number) {
    return double(square(number));
}

console.log(process(5));
````

### Flow

```text
5
 ↓
square()
 ↓
25
 ↓
double()
 ↓
50
```

One function can use the result of another function.

## 19. Higher-Order Functions — Introduction

````javascript
function processNumber(number, operation) {
    return operation(number);
}

function square(number) {
    return number * number;
}

console.log(processNumber(5, square));
````

````javascript
console.log(
    processNumber(5, n => n * 2)
);
````

A **higher-order function** is a function that receives another function as an argument or returns a function.

This idea becomes very important when we learn JavaScript array methods such as `map()`, `filter()`, and `reduce()`.
