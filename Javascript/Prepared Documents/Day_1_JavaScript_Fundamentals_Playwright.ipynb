## 4. Variables

### Definition
A **variable** is a named reference used to store a value.

Example:

```javascript
let username = "admin";
```

Here:
- `username` → variable name
- `"admin"` → value

Think of it as:

```text
username
   ↓
"admin"
```

````typescript
let username = "admin";
console.log(username);
````

### Why use variables?

Instead of repeating values:

```javascript
console.log("admin");
console.log("admin");
console.log("admin");
```

we can store the value once:

```javascript
let username = "admin";

console.log(username);
console.log(username);
console.log(username);
```

This makes code easier to maintain.

````typescript
let username = "admin";

console.log(username);
console.log(username);
console.log(username);
````

## 5. `let` and `const`

### `let`

Use `let` when the value may change.

````typescript
let browser = "Chrome";
console.log(browser);

browser = "Firefox";
console.log(browser);
````

### `const`

Use `const` when the variable should not be reassigned.

For modern JavaScript:

> **Use `const` by default. Use `let` when the value needs to change.**

`var` is an older declaration style. You should recognize it when you see existing code, but prefer `let` and `const` for new code.

````typescript
const application = "Amazon";
console.log(application);
````

## 11. Assignment Operator

The `=` operator assigns a value to a variable.

```javascript
let username = "admin";
```

It means:

> Assign `"admin"` to `username`.

It does not mean mathematical equality.

````typescript
let count = 10;
console.log(count);

count = 20;
console.log(count);
````

## 12. Comparison Operators

Comparison operators compare values.

Important operators:

```text
>    greater than
<    less than
>=   greater than or equal
<=   less than or equal
===  strictly equal
!==  strictly not equal
```

````typescript
const age = 25;

console.log("age > 18:", age > 18);
console.log("age < 18:", age < 18);
console.log("age === 25:", age === 25);
console.log("age !== 30:", age !== 30);
````

## 13. `==` vs `===`

Prefer `===` and `!==` in modern JavaScript.

```javascript
10 == "10"   // true
10 === "10"  // false
```

Why?

`===` checks both the **value and the type**.

- `10` → number
- `"10"` → string

````typescript
console.log("10 == \"10\":", 10 == "10");
console.log("10 === \"10\":", 10 === "10");
````

## 14. Logical Operators

Three important logical operators:

| Operator | Meaning |
|---|---|
| `&&` | AND |
| ` || ` | OR |
| `!` | NOT |

### AND

Both conditions must be true.

### OR

At least one condition must be true.

### NOT

Reverses a Boolean value.

````typescript
const usernameEntered = true;
const passwordEntered = true;

console.log("AND:", usernameEntered && passwordEntered);

const isAdmin = false;
const isManager = true;
console.log("OR:", isAdmin || isManager);

const isLoggedIn = true;
console.log("NOT:", !isLoggedIn);
````
