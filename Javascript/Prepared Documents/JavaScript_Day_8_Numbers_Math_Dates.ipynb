# Part 1 — Numbers

JavaScript has a `number` type for both integers and decimal values.

```javascript
let age = 25;
let price = 999.99;
let temperature = -5;
```

Unlike some languages, JavaScript does not have separate everyday types such as `int` and `float`.


````javascript
let age = 25;
let price = 999.99;
let temperature = -5;

console.log(age);
console.log(price);
console.log(temperature);

console.log(typeof age);
console.log(typeof price);
````

# 5. `Number()`

`Number()` converts a value to a number.

It is stricter than `parseInt()` and `parseFloat()` for strings containing extra characters.


````javascript
console.log(Number("100"));
console.log(Number("99.50"));
console.log(Number("100px"));
````

### Output

```text

```


# 6. `parseInt()` vs `parseFloat()` vs `Number()`

| Function | Example | Result |
|---|---|---:|
| `parseInt("25.75", 10)` | integer parsing | `25` |
| `parseFloat("25.75")` | decimal parsing | `25.75` |
| `Number("25.75")` | full numeric conversion | `25.75` |
| `Number("25px")` | invalid numeric string | `NaN` |

### Practical rule

Use `Number()` when the entire input should be a valid number.

Use `parseInt()` when you specifically want an integer.

Use `parseFloat()` when you specifically want a decimal value.

# 9. `isNaN()` vs `Number.isNaN()`

This distinction is important.

`isNaN()` first attempts numeric conversion.

`Number.isNaN()` checks whether the value is already the special `NaN` value.


````javascript
console.log(isNaN("Hello"));
console.log(isNaN("100"));

console.log(Number.isNaN("Hello"));
console.log(Number.isNaN(NaN));
````

### Output

```text
true
false
false
true
```

For modern JavaScript validation, `Number.isNaN()` is usually clearer when you specifically need to test for `NaN`.

# 38. Calculate Difference Between Two Dates

Subtract their millisecond values.


````javascript
let start = new Date("2026-09-01");
let end = new Date("2026-09-21");

let differenceMs = end.getTime() - start.getTime();

let differenceDays = differenceMs / (1000 * 60 * 60 * 24);

console.log(differenceDays);
````

# 39. Calculate Age — Basic Version

A simple year-difference calculation is:

```javascript
currentYear - birthYear
```

But a fully accurate age calculation must also consider whether the birthday has occurred yet this year.


````javascript
let birthYear = 1995;
let currentYear = new Date().getFullYear();

let age = currentYear - birthYear;

console.log(age);
````

### Example output

```text
31
```

This is a simplified calculation.


# 40. More Accurate Age Calculation


````javascript
function calculateAge(birthDate) {
    let today = new Date();
    let birth = new Date(birthDate);

    let age = today.getFullYear() - birth.getFullYear();

    let birthdayThisYear = new Date(
        today.getFullYear(),
        birth.getMonth(),
        birth.getDate()
    );

    if (today < birthdayThisYear) {
        age--;
    }

    return age;
}

console.log(calculateAge("1995-10-15"));
````

### Output

The exact result depends on the day the code is run.

The function accounts for whether the birthday has occurred in the current year.


# 41. Practical Use Case — Product Price Calculator

Combine number conversion, arithmetic, rounding and template literals.


````javascript
let priceInput = "2499.99";
let quantityInput = "3";
let discountInput = "10";

let price = Number(priceInput);
let quantity = Number(quantityInput);
let discount = Number(discountInput);

let subtotal = price * quantity;
let discountAmount = subtotal * discount / 100;
let finalAmount = subtotal - discountAmount;

console.log(`Subtotal: ₹${subtotal.toFixed(2)}`);
console.log(`Discount: ₹${discountAmount.toFixed(2)}`);
console.log(`Final Amount: ₹${finalAmount.toFixed(2)}`);
````

### Output

```text
Subtotal: ₹7499.97
Discount: ₹749.997
Final Amount: ₹6749.97
```

If displaying currency, format the discount too:

```javascript
discountAmount.toFixed(2)
```

so it appears as `₹750.00`.
