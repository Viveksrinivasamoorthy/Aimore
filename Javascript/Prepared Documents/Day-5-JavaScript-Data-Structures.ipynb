## 1. What is a Data Structure?

A **data structure** is a way of organizing and storing data so that we can work with it efficiently.

Instead of creating many separate variables:

```javascript
let student1 = "Arun";
let student2 = "Priya";
let student3 = "Rahul";
```

we can store related values together:

```javascript
let students = ["Arun", "Priya", "Rahul"];
```

### Main structures in today's class

| Data Structure | Use |
|---|---|
| Array | Store multiple values |
| Object | Store related information using key-value pairs |
| Array of Objects | Store multiple records |


# Part 1 — Arrays

## 2. Creating an Array

An array stores multiple values in a single variable.


````javascript
let fruits = ["Apple", "Banana", "Mango"];

console.log(fruits);
````

### Arrays can contain different data types


````javascript
let data = ["Arun", 25, true];

console.log(data);
````

## 3. Array Index

JavaScript arrays start from **index 0**.

```text
Value:    Apple    Banana    Mango
Index:      0        1         2
```


````javascript
let fruits = ["Apple", "Banana", "Mango"];

console.log(fruits[0]);
console.log(fruits[1]);
console.log(fruits[2]);
````

## 4. Changing an Array Element


````javascript
let fruits = ["Apple", "Banana", "Mango"];

fruits[1] = "Orange";

console.log(fruits);
````

## 5. Array Length

The `length` property tells us how many elements are in an array.


````javascript
let fruits = ["Apple", "Banana", "Mango"];

console.log("Number of fruits:", fruits.length);
````

### Using `length` with a loop


````javascript
let fruits = ["Apple", "Banana", "Mango"];

for (let i = 0; i < fruits.length; i++) {
    console.log(fruits[i]);
}
````

## 6. Adding Elements

### `push()`

Adds an element at the end of the array.


````javascript
let fruits = ["Apple", "Banana"];

fruits.push("Mango");

console.log(fruits);
````

### `unshift()`

Adds an element at the beginning of the array.


````javascript
fruits.unshift("Orange");

console.log(fruits);
````

## 7. Removing Elements

### `pop()`

Removes the last element.


````javascript
let fruits = ["Apple", "Banana", "Mango"];

fruits.pop();

console.log(fruits);
````

### `shift()`

Removes the first element.


````javascript
let fruits = ["Apple", "Banana", "Mango"];

fruits.shift();

console.log(fruits);
````

## 8. Common Array Methods

For today's class, focus on:

- `push()`
- `pop()`
- `shift()`
- `unshift()`
- `length`
- `includes()`
- `indexOf()`


````javascript
let fruits = ["Apple", "Banana", "Mango"];

console.log("Contains Mango:", fruits.includes("Mango"));
console.log("Index of Banana:", fruits.indexOf("Banana"));
````

## 9. Looping Through an Array

We can combine arrays with the `for` loop learned in Day 3.


````javascript
let students = ["Arun", "Priya", "Rahul", "Divya"];

for (let i = 0; i < students.length; i++) {
    console.log(students[i]);
}
````

### Practical Example — Marks


````javascript
let marks = [85, 72, 90, 65, 88];

for (let i = 0; i < marks.length; i++) {
    console.log("Mark:", marks[i]);
}
````

## 10. Finding the Total of Array Values

This combines:

- Array
- Loop
- Variable
- Arithmetic


````javascript
let marks = [80, 75, 90, 85, 70];

let total = 0;

for (let i = 0; i < marks.length; i++) {
    total = total + marks[i];
}

console.log("Total:", total);
````

## 11. Finding the Average


````javascript
let marks = [80, 75, 90, 85, 70];

let total = 0;

for (let i = 0; i < marks.length; i++) {
    total = total + marks[i];
}

let average = total / marks.length;

console.log("Total:", total);
console.log("Average:", average);
````

# Part 2 — Objects

## 12. Why Objects?

Suppose we want to store information about one student.

Without an object:

```javascript
let name = "Arun";
let age = 21;
let course = "JavaScript";
let mark = 85;
```

An object groups related information together.


````javascript
let student = {
    name: "Arun",
    age: 21,
    course: "JavaScript",
    mark: 85
};

console.log(student);
````

## 13. Object Key-Value Pairs

Example:

```javascript
let student = {
    name: "Arun",
    age: 21,
    course: "JavaScript"
};
```

The left side is the **key/property**.

The right side is the **value**.

```text
name   → Arun
age    → 21
course → JavaScript
```


````javascript
let student = {
    name: "Arun",
    age: 21,
    course: "JavaScript"
};

console.log(student);
````

## 14. Accessing Object Values

### Dot notation


````javascript
console.log(student.name);
console.log(student.age);
console.log(student.course);
````

### Bracket notation


````javascript
console.log(student["name"]);
console.log(student["age"]);
````

## 15. Updating Object Values


````javascript
let student = {
    name: "Arun",
    age: 21
};

student.age = 22;

console.log(student);
````

## 16. Adding New Properties


````javascript
let student = {
    name: "Arun",
    age: 21
};

student.city = "Chennai";

console.log(student);
````

## 17. Deleting Properties


````javascript
delete student.city;

console.log(student);
````

# Part 3 — Arrays + Objects

## 18. Array of Objects

An **array of objects** is commonly used to store multiple real-world records.


````javascript
let students = [
    {
        name: "Arun",
        age: 21,
        mark: 85
    },
    {
        name: "Priya",
        age: 22,
        mark: 92
    },
    {
        name: "Rahul",
        age: 20,
        mark: 76
    }
];

console.log(students);
````

### Accessing data inside an array of objects


````javascript
console.log(students[0].name);
console.log(students[1].mark);
console.log(students[2].age);
````

## 19. Looping Through an Array of Objects


````javascript
for (let i = 0; i < students.length; i++) {
    console.log(students[i].name);
}
````

### Print name and mark


````javascript
for (let i = 0; i < students.length; i++) {
    console.log(
        students[i].name,
        students[i].mark
    );
}
````

## 20. Practical Use Case — Student Management

Display students who scored 80 or above.


````javascript
let students = [
    { name: "Arun", mark: 85 },
    { name: "Priya", mark: 92 },
    { name: "Rahul", mark: 65 },
    { name: "Divya", mark: 78 }
];

for (let i = 0; i < students.length; i++) {

    if (students[i].mark >= 80) {
        console.log(students[i].name + " - Excellent");
    }
}
````

## 21. Practical Use Case — Shopping Cart

Calculate the total price of all products.


````javascript
let cart = [
    { name: "Laptop", price: 50000 },
    { name: "Mouse", price: 800 },
    { name: "Keyboard", price: 1500 }
];

let total = 0;

for (let i = 0; i < cart.length; i++) {
    total = total + cart[i].price;
}

console.log("Cart Total:", total);
````

## 22. Practical Use Case — Employee Records

Find all employees who work in the IT department.


````javascript
let employees = [
    {
        name: "Arun",
        department: "IT",
        salary: 50000
    },
    {
        name: "Priya",
        department: "HR",
        salary: 45000
    },
    {
        name: "Rahul",
        department: "IT",
        salary: 60000
    }
];

for (let i = 0; i < employees.length; i++) {

    if (employees[i].department === "IT") {
        console.log(employees[i].name);
    }
}
````

## 23. Practical Use Case — Product Inventory

Check whether each product is available.


````javascript
let products = [
    {
        name: "Laptop",
        price: 55000,
        stock: 10
    },
    {
        name: "Phone",
        price: 25000,
        stock: 0
    },
    {
        name: "Tablet",
        price: 18000,
        stock: 5
    }
];

for (let i = 0; i < products.length; i++) {

    if (products[i].stock > 0) {
        console.log(products[i].name + " is available");
    } else {
        console.log(products[i].name + " is out of stock");
    }
}
````

# Part 4 — Nested Arrays

An array can contain other arrays.

This is useful for tables, matrices, grids, and similar data.


````javascript
let matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
];

console.log(matrix[0][0]);
console.log(matrix[1][2]);
console.log(matrix[2][1]);
````

### Nested loops

This combines the nested loops learned in Day 3 with nested arrays.


````javascript
for (let i = 0; i < matrix.length; i++) {

    for (let j = 0; j < matrix[i].length; j++) {
        console.log(matrix[i][j]);
    }

}
````
