````typescript
console.log("Step 1")
for (i=1;i<=20;i+=1)
{
console.log(i)
}
console.log("Step 2")
for (i=1;i<=10;i+=2)
{
console.log(i)
}
console.log("Step 3")
for (i=1;i<=10;i+=3)
{
console.log(i)
}
// step -> 
// if step is 2: it always keep the first value (i) and skip the next values
// if step is 3: it always keep the first value (i) and skip the next two values
// if step is 4: it always keep the first value (i) and skip the next three values

// break -> come out from the loop and stop the loop execution
// continue -> skip the current iteration and move to the next iteration () stays in the loop)
````

````typescript
/* # 10. `while` Loop

Use `while` when repetition depends mainly on a condition. 

## Important

Update the value used by the condition, otherwise you may create an infinite loop.
*/
````

````typescript
let i = 1;

while (i <= 5) {
    console.log(i);
    i++;
}
````
