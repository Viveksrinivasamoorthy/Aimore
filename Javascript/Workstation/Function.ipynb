````typescript
/*
scope vs global variable
 The variable `name` exists only inside `employee()`. Trying to use it outside that function would cause an error.

### Simple rule

- **Global variable** → available from a wider scope
- **Local variable** → available only inside its scope

Prefer keeping variables in the smallest scope where they are needed.
*/
let company = "ABC";

function employee() {
    let name = "Arun";

    console.log("Inside function - company:", company);
    console.log("Inside function - name:", name);
}

employee();

console.log("Outside function - company:", company);
````

````typescript
/*
## Arrow Functions as Callback Functions

A function passed to another function is a callback. Arrow functions are commonly used for callbacks.
````

````typescript
let nums = [1,2,3,4,5];
nums.forEach(number => console.log(number));
````

````typescript
let nums = [1,2,3,4,5];
nums.forEach(number => console.log(number));
let doubled = nums.map(number => number * 2);
let large = nums.filter(number => number > 3);
let found = nums.find(number => number === 4);
let total = nums.reduce((sum, number) => sum + number, 0);
console.log(doubled, large, found, total);
````

````typescript
// calculate array sum

function calculateSum(numbers) {
    let sum = 0;

    for (let i = 0; i < numbers.length; i++) {
        sum += numbers[i];
    }

    return sum;
}

let numbers = [10, 20, 30, 40];

console.log("Sum:", calculateSum(numbers));
````

````typescript
// count even number
function countEvenNumbers(numbers) {
    let count = 0;

    for (let number of numbers) {
        if (number % 2 === 0) {
            count++;
        }
    }

    return count;
}

console.log(
    countEvenNumbers([10, 15, 20, 23, 30, 35])
);
````
