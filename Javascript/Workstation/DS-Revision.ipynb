```typescript
let marksForMax = [78, 92, 65, 88, 72,55,100];

// highest? 
// lowest?

// trenary? 

// let variable = condition ? true: false
let mark = marksForMax.reduce(
//    (highest, mark) => mark > highest ? mark : highest,   marksForMax[0]    
    (a, b) => a>b  ? a : b,   marksForMax[0]    
);
console.log("highest a>b :", mark);

mark = marksForMax.reduce(
//    (highest, mark) => mark > highest ? mark : highest,   marksForMax[0]    
    (a, b) => a<b  ? a : b,   marksForMax[0]    
);
console.log("lowest a<b :", mark);
```
