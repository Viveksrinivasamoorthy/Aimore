```typescript
/*
JavaScript Map() — Detailed Explanation

Map is a built-in JavaScript data structure used to store data in key → value pairs.

Think of it like a dictionary:

Key       Value
--------------------
101   →   "Arun"
102   →   "Priya"
103   →   "Rahul"

The important difference from a normal JavaScript object is that a 
Map is specifically designed for key-value data and its keys can be any JavaScript value.
Map will not allow duplicates


Difference btw:


Feature             MAP                                         Object
Key  Order          Preserves insertion order during iteration.	Mostly insertion order, but integer keys are sorted first.
Size Retrieval      Direct via .size property                   Manual calculation using Object.keys(obj).length
Performance         Optimized for frequent addition             Optimized for static structures and fixed key setups.
                    and removal of pairs.
Iteration           Directly iterable via                       Not directly iterable; requires Object.keys(), values(), or entries()
                    for...of loops or .forEach().            
```

```typescript
// for (initializatio;condition;in)
// value,key value, index
// object_name.key 
// i=0; i< length()

const fruits = new Map([
  ["apple", 50],
  ["banana", 20]
]);


// for ( let i=0;i<obj.length();i++)
// for (let [i,j] of map_name)

// Loop over key-value pairs
for (let [i, j] of fruits) 
{
    console.log("Fruit:",i," Price:$",j)
}
```

```typescript


const fruits = new Map([
  ["apple", 50],
  ["banana", 20]
]);

// Loop over key-value pairs
for (const [fruit, price] of fruits) {
  console.log(`${fruit}: $${price}`);
}

// Using forEach (Note: value comes first, key comes second)
fruits.forEach((price, fruit) => {
  console.log(`${fruit}: $${price}`);
});
```
