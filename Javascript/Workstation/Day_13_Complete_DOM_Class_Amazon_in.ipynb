## 1. HTML vs DOM

HTML describes the page structure. The browser parses it and creates a live object model called the DOM.

```text
HTML → Browser parses HTML → DOM tree → JavaScript interacts with it
```

Example HTML:

```html
<form>
  <input id="searchBox" type="text">
  <button>Search</button>
</form>
```

Conceptual tree:

```text
document
└── html
    └── body
        └── form
            ├── input
            └── button
```

**Teaching point:** JavaScript works with DOM objects, not directly with the HTML file as plain text.

## 4. CSS selectors

| Selector | Meaning | Example |
|---|---|---|
| `input` | HTML tag | `document.querySelector("input")` |
| `#searchBox` | ID | `document.querySelector("#searchBox")` |
| `.nav-input` | Class | `document.querySelector(".nav-input")` |
| `[name="email"]` | Attribute | `document.querySelector('[name="email"]')` |
| `input[type="text"]` | Tag + attribute | `document.querySelector('input[type="text"]')` |
| `.card a` | Anchor inside `.card` | `document.querySelector(".card a")` |
| `.btn.primary` | Element with both classes | `document.querySelector(".btn.primary")` |

`#` means ID; `.` means class.

`getElementById()` takes the ID without `#`:

```javascript
document.getElementById("twotabsearchtextbox"); // correct
document.getElementById("#twotabsearchtextbox"); // incorrect
```

Prefer stable, observed selectors. Avoid long, fragile chains of nested elements.

## 5. `querySelector()` vs `querySelectorAll()`

`querySelector()` returns the first match or `null`:

```javascript
const firstInput = document.querySelector("input");
console.log(firstInput);
```

`querySelectorAll()` returns a static NodeList containing all matches:

```javascript
const inputs = document.querySelectorAll("input");
console.log(inputs);
console.log(inputs.length);
console.log(inputs[0]);
```

Remember:

```text
querySelector()     → first match or null
querySelectorAll()  → all matches in a NodeList
```

Loop through all matches:

```javascript
document.querySelectorAll("a").forEach(link => {
  console.log(link.textContent);
});
```

A `querySelectorAll()` NodeList is static; run the query again to get a fresh list after the DOM changes.

## 6. Reading text: `textContent`, `innerText`, `innerHTML`

Read text:

```javascript
const heading = document.querySelector("h1");
if (heading) console.log(heading.textContent);
```

- `textContent`: text in the DOM, including text not necessarily visible.
- `innerText`: text as rendered/visible, affected by layout.
- `innerHTML`: markup inside an element.

```javascript
element.textContent = "<b>Hello</b>"; // displays the angle-bracket text
element.innerHTML = "<b>Hello</b>";   // interprets markup and displays bold Hello
```

Use `textContent` for ordinary text. Assigning untrusted strings to `innerHTML` can introduce security risks.

## 7. Input values

For form controls such as `<input>`, `<textarea>`, and `<select>`, use `.value`.

```javascript
const searchBox = document.querySelector("#twotabsearchtextbox");
if (searchBox) {
  console.log(searchBox.value);
}
```

Amazon practice, after verifying the selector:

```javascript
if (searchBox) {
  searchBox.value = "wireless mouse";
  console.log(searchBox.value);
}
```

Directly assigning `.value` may not trigger the same application behavior as real typing. Modern sites can listen for input events and maintain internal state. For real automation, later use browser automation APIs such as Playwright's `fill()` and `click()`.

## 8. Attributes and properties

Attributes include `id`, `class`, `name`, `type`, `href`, `src`, and `placeholder`.

```javascript
const searchBox = document.querySelector("#twotabsearchtextbox");

if (searchBox) {
  console.log(searchBox.getAttribute("id"));
  console.log(searchBox.getAttribute("name"));
  console.log(searchBox.getAttribute("placeholder"));
}
```

Set or remove an attribute:

```javascript
searchBox?.setAttribute("placeholder", "DOM practice");
searchBox?.removeAttribute("placeholder");
```

Attribute vs property: `getAttribute("value")` reads the markup attribute; `.value` reads the current form value. They can differ after the user edits a field.

## 9. Classes and styles

Use `classList` to manage classes:

```javascript
element.classList.add("active");
element.classList.remove("active");
element.classList.contains("active");
element.classList.toggle("active");
```

Change inline styles:

```javascript
const heading = document.querySelector("h1");
if (heading) {
  heading.style.color = "blue";
  heading.style.backgroundColor = "lightyellow";
  heading.style.fontSize = "32px";
}
```

CSS `background-color` becomes JavaScript `backgroundColor` (camelCase). These are temporary local changes.

## 10. Creating, adding, and removing elements

Create and configure an element:

```javascript
const note = document.createElement("p");
note.textContent = "Added by JavaScript for practice.";
```

Add it to the page:

```javascript
document.body.appendChild(note);
// or: document.body.append(note);
```

Remove it:

```javascript
note.remove();
```

Teaching sequence: **create → configure → insert → inspect → remove**. Practise on a simple page first; remove temporary additions from Amazon.in.

## 11. DOM traversal

Useful properties for navigating around the DOM:

```javascript
element.parentElement
element.children
element.firstElementChild
element.lastElementChild
element.nextElementSibling
element.previousElementSibling
```

Example:

```javascript
console.log(document.body.firstElementChild);
console.log(document.body.children.length);
```

When inspecting an Amazon product card, the title, image, link, and price may be nested inside a container. Traversal helps students understand these relationships, but a clear selector is often more robust than a long chain of parent/child steps.

## 12. Events and `addEventListener()`

An event is something that happens in the browser: `click`, `input`, `change`, `submit`, `keydown`, or `keyup`.

```javascript
const button = document.querySelector("#demoButton");

button?.addEventListener("click", () => {
  console.log("Button clicked");
});
```

Input example on a page containing `#nameInput` and `#output`:

```javascript
const nameInput = document.querySelector("#nameInput");
const output = document.querySelector("#output");

nameInput?.addEventListener("input", () => {
  if (nameInput && output) {
    output.textContent = nameInput.value;
  }
});
```

Events matter because real user interaction can trigger website application logic. Direct property assignment does not always reproduce the same behavior as typing.

## 13. `forEach()`, `map()`, and `filter()` with DOM

Print some Amazon page links:

```javascript
[...document.querySelectorAll("a")]
  .slice(0, 20)
  .forEach((link, index) => {
    console.log(index, link.textContent.trim(), link.href);
  });
```

`querySelectorAll()` returns a NodeList. Convert it to an array with `[...nodeList]` to use array methods:

```javascript
const linkTexts = [...document.querySelectorAll("a")]
  .map(link => link.textContent.trim())
  .filter(text => text !== "");

console.log(linkTexts);
```

This connects the DOM with previously learned array methods and arrow functions.

## 14. Common mistakes and debugging

**Selector returns `null`:** selector may be wrong, element may not exist on this page, or it may not have loaded yet.

**Empty NodeList:** no elements matched at query time.

**Cannot read properties of null:** check before using an element:

```javascript
const title = document.querySelector("#missing");

if (title) {
  title.textContent = "Hello";
} else {
  console.log("Element not found");
}
```

**Wrong property:** use `.value` for form controls and `.textContent` for text.

**Changes disappear:** DevTools edits are local and temporary; refresh usually removes them. The website's server-side data is not changed.

## 17. Final DOM cheat sheet

| Requirement | Method/property |
|---|---|
| Get by ID | `document.getElementById()` |
| First match | `document.querySelector()` |
| All matches | `document.querySelectorAll()` |
| Read/change text | `element.textContent` |
| Read/change form value | `element.value` |
| Read/change markup | `element.innerHTML` |
| Change inline style | `element.style.property` |
| Add/remove/check class | `element.classList` |
| Read attribute | `getAttribute()` |
| Set/remove attribute | `setAttribute()` / `removeAttribute()` |
| Create element | `document.createElement()` |
| Insert element | `append()` / `appendChild()` |
| Remove element | `element.remove()` |
| Listen for event | `addEventListener()` |
| Traverse DOM | `parentElement`, `children`, sibling properties |

### Central mental model

```text
FIND → INSPECT → READ / INTERACT → VERIFY
```

Teach students to reason through each step rather than memorize methods in isolation.
