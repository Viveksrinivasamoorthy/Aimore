```typescript
/*
13. Find the Second Largest Number

Given:

let numbers = [10, 45, 23, 89, 45, 67, 89, 12];

Write a program to find the second largest unique number.

Expected:
67

Do not simply sort the array and take an index.

Your solution should handle:

[10, 10, 10, 5]

appropriately.

*/

let numbers = [10, 45, 23, 89, 45, 67, 89, 12];

uniquenumber = [...new Set(numbers)]
console.log(uniquenumber)

uniquenumber.sort((a,b) => b-a)
console.log(uniquenumber)
console.log("Second largest number:",uniquenumber[1])
```
