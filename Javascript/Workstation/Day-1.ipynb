````typescript
/* ## 7. Data Types

A **data type** describes the kind of value stored in a variable.

Important types for today:

1. String
2. Number
3. Boolean
4. Undefined
5. Null
*/
````

````typescript
/*
### String

A string represents text.

```javascript
const username = "admin";
const browser = "Chrome";
```

Real Playwright example:

```javascript
const username = "testuser";
const password = "Password123";

await page.locator("#username").fill(username);
await page.locator("#password").fill(password);
```
*/
````

````typescript
/*
### Number

Numbers are used for numeric values.

```javascript
const age = 25;
const price = 100;
const quantity = 5;
```
*/

const price = 500;
const quantity = 2;
const total = price * quantity;

console.log("Total:", total);
````

````typescript
/*
### Boolean

A Boolean has only two possible values:

```javascript
true
false
```

Example:

```javascript
const isLoggedIn = true;
const isAdmin = false;
```
*/
const isLoggedIn = true;
const isAdmin = false;

console.log("Logged in:", isLoggedIn);
console.log("Admin:", isAdmin);
````

````typescript
/*
### Undefined

A variable that exists but has not been assigned a value contains `undefined`.

```javascript
let username;
console.log(username);
```
*/

let unassignedUsername;
console.log(unassignedUsername);
````

````typescript
/*
### Null

`null` is an intentional empty/no-value value.

```javascript
const selectedUser = null;
console.log(selectedUser);
```
*/

const selectedUser = null;
console.log(selectedUser);
````

````typescript
/*
## 9. Input → Processing → Output

A basic programming pattern is:

```text
Input
  ↓
Processing
  ↓
Output
```

Example:

```javascript
const price = 100;
const quantity = 2;

const total = price * quantity;

console.log(total);
```

Input:
- `price = 100`
- `quantity = 2`

Processing:
- `100 × 2`

Output:
- `200`
*/

const price = 100;
const quantity = 2;
const total = price * quantity;

console.log("Total:", total);
````

````typescript
/*
## 10. Arithmetic Operators

| Operator | Meaning |
|---|---|
| `+` | Addition |
| `-` | Subtraction |
| `*` | Multiplication |
| `/` | Division |
| `%` | Remainder |
*/

const a = 10;
const b = 3;

console.log("Addition:", a + b);
console.log("Subtraction:", a - b);
console.log("Multiplication:", a * b);
console.log("Division:", a / b);
console.log("Remainder:", a % b);
````

````typescript
/*
## 11. Assignment Operator

The `=` operator assigns a value to a variable.

```javascript
let username = "admin";
```

It means:

> Assign `"admin"` to `username`.

It does not mean mathematical equality.
*/
````

````typescript
/*
## 12. Comparison Operators

Comparison operators compare values.

Important operators:

```text
>    greater than
<    less than
>=   greater than or equal
<=   less than or equal
===  strictly equal
!==  strictly not equal
```
*/

const age = 25;

console.log("age > 18:", age > 18);
console.log("age < 18:", age < 18);
console.log("age === 25:", age === 25);
console.log("age !== 30:", age !== 30);
````

````typescript
/*
## 13. `==` vs `===`

Prefer `===` and `!==` in modern JavaScript.

```javascript
10 == "10"   // true
10 === "10"  // false
```

Why?

`===` checks both the **value and the type**.

- `10` → number
- `"10"` → string
*/

console.log("10 == :\"10\":", 10 == "10");
console.log("10 === \"10\":", 10 === "10");
````
