````typescript
/*
## 9. Input → Processing → Output

A basic programming pattern is:

```text
Input
  ↓
Processing
  ↓
Output
```

Example:

```javascript
const price = 100;
const quantity = 2;

const total = price * quantity;

console.log(total);
```

Input:
- `price = 100`
- `quantity = 2`

Processing:
- `100 × 2`

Output:
- `200`
*/

const price = 100;
const quantity = 2;
const total = price * quantity;

console.log("Total:", total);
````

````typescript
/*
## 10. Arithmetic Operators

| Operator | Meaning |
|---|---|
| `+` | Addition |
| `-` | Subtraction |
| `*` | Multiplication |
| `/` | Division |
| `%` | Remainder |
*/

const a = 10;
const b = 3;
const addition = a + b;
const subtraction = a - b;
const multiplication = a * b;
const division = a / b;
const remainder = a % b;
console.log("Addition:", addition);
console.log("Subtraction:", subtraction);
console.log("Multiplication:", multiplication);
console.log("Division:", division);
console.log("Remainder:", remainder);
````

````typescript
/*
## 12. Comparison Operators

Comparison operators compare values. Return true or false

Important operators:

```text
>    greater than
<    less than
>=   greater than or equal
<=   less than or equal
===  strictly equal
!==  strictly not equal
```
*/

// const age = 25;

console.log("age > 18:", age > 18);
console.log("age < 18:", age < 18);
console.log("age === 25:", age === 25);
console.log("age !== 30:", age !== 30);


let xy = "10";
console.log(typeof xy); // string
console.log("xy === 10:", xy === 10); // false, because types are different
console.log("xy == 10:", xy == 10); // true, because == does type coercion

````

````typescript
/*
## 13. `==` vs `===`

Prefer `===` and `!==` in modern JavaScript.

```javascript
10 == "10"   // true
10 === "10"  // false
```

Why?

`===` checks both the **value and the type**.

- `10` → number
- `"10"` → string
*/

console.log("10 == :\"10\":", 10 == "10");
console.log("10 === \"10\":", 10 === "10");
````

````typescript
/*
## 14. Logical Operators

Three important logical operators:

| Operator | Meaning |
|---|---|
| `&&` | AND |
| ` || ` | OR |
| `!` | NOT |

### AND

Both conditions must be true.

### OR

At least one condition must be true.

### NOT

Reverses a Boolean value.
*/
````

````typescript
const usernameEntered = true;
const passwordEntered = true;

console.log("AND:", usernameEntered && passwordEntered);

const isAdmin = false;
const isManager = true;
console.log("OR:", isAdmin || isManager);

let isLoggedin = true;
console.log("NOT:", !isLoggedin);


isLoggedin = false;
console.log("NOT:", !isLoggedin);
````
