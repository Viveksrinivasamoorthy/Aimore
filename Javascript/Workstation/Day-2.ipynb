````typescript
/*
# Part 2 — `if` Statement

An `if` statement allows JavaScript to make a decision.

### Syntax

```javascript
if (condition) {
    // code
}
```
*/
````

````typescript
let age = 20;

if (age >= 18) {
    console.log("You are an adult"); // The condition produces `true`, so JavaScript executes the code inside `{ }`.
}
````

````typescript
let temperature = 35;

if (temperature > 30) {
    console.log("It is hot");
}
````

````typescript
let marks = 80;

if (marks >= 50) {
    console.log("Pass");
}
````

````typescript
/*
# Part 3 — `if...else`

`else` gives JavaScript a second path when the condition is false.

### Syntax

```javascript
if (condition) {
    // true
} else {
    // false
}
```
*/
````

````typescript
let age = 16;

if (age >= 18) {
    console.log("Adult");
} else {
    console.log("Minor");
}
````

````typescript
let marks = 45;

if (marks >= 50) {
    console.log("Pass");
} else {
    console.log("Fail");
}
````

````typescript
/*
# Part 4 — `else if`

Use `else if` when there are multiple possible conditions.

### Syntax

```javascript
if (condition1) {

} else if (condition2) {

} else {

}
```
*/
````

````typescript
let marks = 85;

if (marks >= 90) {
    console.log("A");
} else if (marks >= 75) {
    console.log("B");
} else if (marks >= 50) {
    console.log("C");
} else {
    console.log("Fail");
}
````

````typescript
/*
# Part 5 — Logical Operators in Conditions

This connects directly to Day 1.

## `&&` — AND

Both conditions must be true.

## `||` — OR

At least one condition must be true.

## `!` — NOT

Reverses a Boolean value.

*/
````

````typescript
let age = 25;

if (age >= 18 && age <= 60) {
    console.log("Working age group");
}
````

````typescript
let day = "Sunday";

if (day === "Saturday" || day === "Sunday") {
    console.log("Weekend");
}
````

````typescript
let isLoggedIn = false;

if (!isLoggedIn) {
    console.log("Please login");
}
````

````typescript
/*
# Part 6 — Nested `if`

A nested `if` is an `if` statement inside another `if`.
*/

let age = 20;
let hasLicense = true;

if (age >= 18) {
    if (hasLicense) {
        console.log("You can drive");
    }
}
````

````typescript
// Sometimes nested conditions can be simplified with logical operators:
let age = 20;
let hasLicense = true;

if (age >= 18 && hasLicense) {
    console.log("You can drive");
}
````

````typescript
/*
# Part 7 — Ternary Operator

The ternary operator is a compact way to write a simple `if...else`.

### Normal version
*/
````

````typescript
let age = 20;

if (age >= 18) {
    console.log("Adult");
} else {
    console.log("Minor");
}
````

````typescript
// ### Ternary version

let age = 20;

let result = age >= 18 ? "Adult" : "Minor";

console.log(result);
````

````typescript
/*

### Syntax

```javascript
condition ? valueIfTrue : valueIfFalse
```

Use the ternary operator mainly for short, simple decisions.


# ⏸️ 10-Minute Break / Individual Challenge

Given:

```javascript
let marks = 72;
```

Create a program with these rules:

- `90–100` → A
- `75–89` → B
- `50–74` → C
- Below `50` → Fail

*/
````

````typescript
let score = 82;

if (score >= 90 && score <= 100) {
  console.log("A");
} else if (score >= 75) {
  console.log("B");
} else if (score >= 50) {
  console.log("C");
} else {
  console.log("Fail");
}
````

````typescript
// using ternary operator
let score = 82;

let result = (score >= 90 && score <= 100) ? "A"
           : score >= 75 ? "B"
           : score >= 50 ? "C"
           : "Fail";

console.log(result);
````

````typescript
/*
# Part 9 — `for` Loop

The `for` loop is the main loop for Day 2.

### Syntax

```javascript
for (initialization; condition; update) {
    // code
}
```
*/
````

````typescript
for (let i = 1; i <= 5; i++) {
    console.log(i);
}
````

````typescript
/*
## `for` Loop Examples

### Print 1–10
*/

for (let i = 1; i <= 10; i++) {
    console.log(i);
}
````

````typescript
// print even numbers from 1 to 20

for (let i = 2; i <= 10; i += 2) {
    console.log(i);
}
````

````typescript
// countdown from 5 to 1
for (let i = 5; i >= 1; i--) {
    console.log(i);
}
````
