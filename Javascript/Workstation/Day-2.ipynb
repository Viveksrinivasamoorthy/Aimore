````typescript
/*
# Part 7 — Ternary Operator

The ternary operator is a compact way to write a simple `if...else`.

### Normal version
*/
````

````typescript
let age = 20;
let result = ""

if (age >= 18) 
{
    result = "Adult";  
} 
else 
{
    result = "Minor";    
}
console.log(result);
````

````typescript
// ### Ternary version

let age = 20;

let result = age >= 18 ? "Adult" : "Minor";

console.log(result);
````

````typescript
// using ternary operator

 

let score = 82;


let result = (score >= 90 && score <=100) ? "A" : (score >=75) ? "B" : (score>=50) ? "C" : "fail"

console.log(result);
````

````typescript
/*
## 8. `switch` Statement

`switch` is useful when comparing one value against several fixed choices.

switch(value) {
case value:
    statement;
    break;
    

*/
````

````typescript
let day = 1;

switch (day) {
    case 1:
        console.log("Monday");
        break;

    case 2:
        console.log("Tuesday");
        break;

    case 3:
        console.log("Wednesday");
        break;

    default:
        console.log("Invalid day");
}
````

````typescript
/*
### Why `break`?

Without `break`, JavaScript can continue executing the following cases after a match. `default` runs when no case matches.
*/
````

````typescript
/*
## `for` Loop Examples

### Print 1–10
*/

for (let i = 1; i <= 10; i++) {
    console.log(i);
}
````

````typescript
// print even numbers from 1 to 20

for (let i = 2; i <= 10; i += 2) {
    console.log(i);
}
````

````typescript
// countdown from 5 to 1
for (let i = 5; i >= 1; i--) {
    console.log(i);
}
````
