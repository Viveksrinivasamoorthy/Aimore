```python
import re


text = "The quick brown fox jumps over the lazy dog."
pattern = "fox"

match = re.search(pattern, text)

if match:
    print(f"Found '{pattern}' at index: {match.start()} end: {match.end()}")
else:
    print("Pattern not found")
```

```python
data = "Python_3! 2026"

# \d (Digit) vs \D (Non-digit)
print(re.findall(r"\d", data))
print(re.findall(r"\d+", data))
print(re.findall(r"\D", data))
print(re.findall(r"\D+", data))


```

```python
# \w (Word) vs \W (Non-word)
data = "Python_3! 2026, 2027"

print(re.findall(r"\w+", data)) 
print(re.findall(r"\w", data)) 
print(re.findall(r"\W", data)) 
print(re.findall(r"\W+", data)) 
```

```python
data = "Python_3! 2026, 2027"
# \s (Space) vs \S (Non-space)
print(re.findall(r"\s", data)) 
print(re.findall(r"\S+", data)) 
```

```python
# \b (Boundary): 
data = "Python_3! in 20 26 binary"
print(re.findall(r"\bin\b", data))
```

```python
data = "Python_3! 2026"
print(re.findall(r"[0-9]", data))
print(re.findall(r"[^0-9]", data))
print(re.findall(r"\D+", data))
```

```python
# Extracting Domain Names from Emails
import re
emails = "contact@google.com, support@python.org, hr@startup.io"
# Pattern: Word characters before @ and word characters + . + word characters after
print(re.findall(r"\w+",emails))
print(re.findall(r"\W+",emails))
print(re.findall(r"\.",emails))
print(re.findall(r"\w+\.\w+",emails))
print(re.findall(r"@(\w+\.\w+)",emails))
print(re.findall(r"(\w+\.\w+)",emails))

# print(re.findall(r"\w+\.\w+",emails))
# print(re.findall(r"@\w+.\w+",emails))
# print(re.findall(r"\w+\w+",emails))
# print(re.findall(r"",emails))


# \. vivek@gmail.com, vivek1@gmail.com -> vivek@gmail
 
```

```python
# Mastering Sequences
text = "The year was 2000, and the event happened at midnight."
word_boundary = re.findall(r"\bat\b", text)
print(word_boundary)
whitespace_match = re.findall(r"year\swas", text)
print(whitespace_match)

# year_match = re.findall(r"\b\b",text)
year_match = re.findall(r"\b\d{4}\b",text)
print(year_match)
```

```python
# Cleaning Strings (Replacing Patterns)
raw_input = "User_Name! @#123"
# Replace anything that isn't a letter or digit with ""
clean_name = re.sub(r"[^a-zA-Z0-9]", "", raw_input)  # repalce("!@#","")
clean_name = re.sub(r"[a-zA-Z0-9]", "", raw_input)   # repalce("UserName123","")
print(clean_name)
```

```python
#Splitting Text by Multiple Delimiters
data = "Python,Java;C++|Ruby$C SQL"
languages = re.split(r"[, ;|$]", data)

print(f"Language List: {languages}")
```

```python
import re
text = "Found 123 in the text, and 456 again 365."
text = "email contact@google.com, support@python.org, hr@startup.io"
pattern = r"@(\w+\.\w+)"

# Iterate through match objects
for match in re.finditer(pattern, text):
    print(f"Domain Name '{match.group()}' at index {match.start()}-{match.end()}")
```
