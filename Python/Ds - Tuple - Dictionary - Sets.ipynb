```python
# tuple()
# it is like list BUT can't change - It is immutable
# faster than list
# also ordered and allows duplciate
```

```python
tuple1 = (10,20,30,40,50)
tuple2 = (50,60,70,80,90)
print(tuple1[0])
print(tuple1[-1])
print(tuple1[:])
print(tuple1[::2])
print(tuple1[::-1])
tuple3 = tuple1+tuple2
print(tuple1+tuple2)
print(tuple3)

print(len(tuple2))
print(tuple3.count(50))
print(min(tuple3))
print(max(tuple3))
print(sum(tuple3))
tuple4 = (10,20,30,0)
print(all(tuple4))
tuple5 = (0,False,None,"")
print(any(tuple5))
tuple6 = (0,False,None,"",[0])
print(any(tuple6))
tuple7 = (0,False,None,"",[None])
print(any(tuple7))
```

```python

tuple8 = 10
print(tuple8)
print(type(tuple8))
tuple8 = (10)
print(tuple8)
print(type(tuple8))
tuple8 = ("vivek")
print(tuple8)
print(type(tuple8))

tuple8 = (10,)
print(tuple8)
print(type(tuple8))
tuple8 = ("vivek",)
print(tuple8)
print(type(tuple8))
```

```python
tuple1 = (10,20,30,40,50,50,60,70,80,90)
print(tuple1)
list1 = list(tuple1)
print(type(list1))
print(list1)
list1.remove(50)
# print(list1)
tuple2 = tuple(list1)
print(tuple2)
```
