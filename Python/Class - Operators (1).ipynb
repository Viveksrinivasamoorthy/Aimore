```python
# arithmetic operator
# +, -, *, /, %, //, **

a=11
b=3
# print("a + b | ",a," + ",b," = ",a+b)
print(f"a + b | {a} + {b} = {a+b}")
print(f"a - b | {a} - {b} = {a-b}")
print(f"a / b | {a} / {b} = {a/b}")
print(f"a * b | {a} * {b} = {a*b}")
print(f"a // b | {a} // {b} = {a//b}")
print(f"a ** b | {a} ** {b} = {a**b}")
print(f"a % b | {a} % {b} = {a%b}")
# print(f"Value of a is:{ a     }")
# a + b | 10 + 5 = 15
```
