```python
# sets {} | set() -> conversion from list to set

-> It is an inbuild Data structure which does not allow duplicates.
-> it is unordered
-> It is Mutable

union -> eliminates duplicates and orders result in ascending
intersection -> shows matched data between the data sets
minus -> shows unmatched data from 1st data set
```

```python
numbers = {1, 2, 3, 4, 5,5,4,3,2,1}
print(numbers)  
a= {1,2,3}
b = {3,4,5}
print("--------------")
print(a.union(b))
print(a.intersection(b))
print(a.difference(b))


print("--------------")

print(a|b)
print(a&b)
print(a-b)

print("--------------")
a.add(6)
print(a)
a.remove(2)
print(a)
print(len(a))
```

```python
# dictionary - {} - key,value

subjects = ["English", "Math", "Science"];
marks = [85, 90, 78];



student_grades = {"English": 85, "Math": 90, "Science":78}

print(type(subjects)) 
print(type(marks)) 
print(type(student_grades)) 

# get(), keys() values() items() update() pop() popitem() clear()  setdefault() - returns the value for the specified key if key is in dictionary, else default_value. If default_value is not provided, it returns None.

# get(key,default_value) - returns the value for the specified key if key is in dictionary, else default_value. If default_value is not provided, it returns None.
print(student_grades.get("Math"))
print(student_grades.get("English"))
print(student_grades.get("Tamil"))
print(student_grades.get("Tamil","Not Found"))

print(student_grades.keys())
print(student_grades.values())
print(student_grades.items())


# isnull(age,0)
# coalesce(name,"Unknown")

# update() - updates the dictionary with the specified key-value pairs. If the key already exists, its value is updated; if the key does not exist, a new key-value pair is added.
student_grades.update({"Math": 100})
print(student_grades.get("Math"))
print(student_grades.items())
student_grades.update({"Tamil": 95})
print(student_grades.items())



# pop(key) - removes the specified key and returns its value. If the key does not exist, it raises a KeyError.
print(student_grades.pop("Tamil"))
print(student_grades)
#popitem() - removes and returns the last inserted key-value pair as a tuple. If the dictionary is empty, it raises a KeyError.

print(student_grades.popitem())
print(student_grades)


student_grades.clear()
print(student_grades)


student_grades = {"History":88}
print(student_grades)
student_grades.setdefault("Math", 90)
print(student_grades)
student_grades.update({"Math": 100,"English": 85,"Science": 78,"Geography": 92})
print(student_grades)
# print(math_marks)

# uniqueness between update and setdefault
# if the key exists, value get updated 
# if the key does not exist, value is added

# difference with syntax - update() takes a dictionary as an argument, while setdefault() takes a key and a default value as arguments.

# creating dict using two lists - dict(zip(keys, values)) - creates a dictionary by zipping two lists together, where the first list contains the keys and the second list contains the corresponding values.

#dict(zip(keys,value))

```
