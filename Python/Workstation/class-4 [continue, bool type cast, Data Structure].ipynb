```python
# access method
fruits = ['apple', 'grape', 'orange','apple', 'grape', 'orange']

print(type(fruits)) # <class 'list'>
print(fruits[0]) # apple
print(fruits[1]) # grape
print(fruits[2]) # orange
print(fruits[-1]) # orange

print(fruits[0:2]) # orange
print(fruits[::]) # orange
print(fruits[::-1]) # orange
print(fruits[::1]) # orange
print(fruits[::2]) # orange
```

```python
# update method

fruits = ['apple', 'grape', 'orange']
print(fruits)

fruits[2] = 'banana' # update approach
print(fruits[2])
print(fruits)
```

```python
# append method
fruits = ['apple', 'grape', 'orange']
print(fruits)
fruits.append('banana') # append approach
print(fruits[3])
print(fruits)
```

```python
# Extend		
mark1 = [50, 60, 70]
mark2 = [80, 90, 100]
mark1.extend(mark2) # extend approach
print(mark1)
mark1.extend([800, 900, 1000]) # extend approach
print(mark1)
print(mark2)
```

```python
# Remove		
fruits = ['apple', 'grape', 'orange']
print(fruits)
fruits.remove('grape') # remove approach
print(fruits)
```

```python
# sort
# all values should be of same type
# order of sorting is based on ASCII value
fruits = ['apple', 'grape', 'orange','banana']
fruits.sort()
print(fruits)
fruits = ['apple', 'grape', 'orange','banana',"Z"]
fruits.sort()
print(fruits)
fruits = ['apple', 'grape', 'orange','banana',"0"," ","ZZ"]
fruits.sort()
print(fruits)

fruits = ['apple', 'grape', 'orange','banana',"0"," ","ZZ"]
print(fruits.sort())
print(fruits)
print("--------")
sortedFruits = fruits.sort()
print(sortedFruits)
```

```python
# count() - returns the number of occurrences of a value in the list
fruits = ['apple', 'grape', 'orange','apple',"0"," ","ZZ"]
print(fruits.count('apple')) # 2
```

```python
# insert() - inserts a value at a specific index in the list
fruits = ['apple', 'grape', 'orange','banana']
fruits.insert(2, 'kiwi') # insert approach
print(fruits)
```

```python
# pop() - removes and returns the last value in the list
fruits = ['apple', 'grape', 'orange','banana']
print(fruits.pop()) # banana
print(fruits)
```

```python
# pop(index) - removes and returns the value at a specific index in the list
fruits = ['apple', 'grape', 'orange','banana']
print(fruits)
fruits.pop(1) # remove grape
print(fruits)
```

```python
# sorted() - returns a new sorted list without changing the original list
# sorted funciton has two parameters - 
# first is the list to be sorted and second is the order of sorting (ascending or descending)
# without second parameter - by default ascending order
fruits = ['apple', 'grape', 'orange','banana',"0","Za","ZZ"]
ascSortFruits = sorted(fruits)
print(ascSortFruits)
print(ascSortFruits)

descSortFruits = sorted(fruits, reverse=True)
print(descSortFruits)
print(fruits)
```

```python
# all() - returns True if all the values in the list are True,
# otherwise returns False
# it is like AND  operator
# false values - 0, "", None, False, []

list1 = [1, 2, 3, 4, 5]
list2 = [1, 2, 3, 4, 0]
list3 = [1, 2, 3, 4, [""],[None],[0],[]]
print(all(list1)) # True
print(all(list2)) # False
print(all(list3)) # False
```

```python
# any() - returns True if any of the values in the list are True,
# otherwise returns False
# OR operator
list3 = [0, 0, 0, 0, 0, None,False]
list4 = [0, 0, 0, 0, 1] 
print(any(list3)) # False
print(any(list4)) # True
```
