```python
correct_password = "Abitha123"
attempts = 3

while attempts > 0:
    password = input("Enter your password: ")

    if password == correct_password:
        print("Loggined successfully!")
        break
    else:
        attempts -= 1

        if attempts > 0:
            print("Wrong password!")
            print("Attempts remain:", attempts)
        else:
            print("Account Locked!")
```
