```python
attempts = 0

while attempts < 3:
    password = input("Enter your password: ")

    if password == "Python123":
        print("Correct Password!")
        break
    else:
        print("Wrong Password!")
        attempts += 1

if attempts == 3:
    print("Maximum attempts reached. Access Denied!")
```

```python
scores = []

while True:
    score = int(input("Enter score (max 100): "))

    if score > 100:
        print("Invalid score entered. Stopping input.")
        break

    scores.append(score)

unique_scores = list(set(scores))

print("Valid scores collected:", unique_scores)
```
