```python
# tuple ()
-> A tuple is like List but we can't change it (immutable)
-> It is also ordered and allows duplicates.
-> It is faster than List.
```

```python
fruits = ('apple', 'grape', 'orange','grape')
print(type(fruits))
tuple1= ("vivek",)
print(type(tuple1))
print(tuple1)
```

```python
print(fruits[::1])
print(fruits[::-1])
```

```python
print(fruits.count('grape'))
print(len(fruits))
print(fruits.index('grape'))
tuple1 = ('apple', 'grape', 'orange')
tuple2 = ('banana', 'kiwi', 'mango')
tuple3 = tuple1 + tuple2
print(tuple3)
```

```python
# list() tuple()
# string to int = int(string)
# int to string = str(int)
# list to tuple = tuple(list)
# tuple to list = list(tuple)

fruits_list = ['apple', 'grape', 'orange']
print(fruits_list)
print(type(fruits_list))
fruit_tuple = tuple(fruits_list)
print(fruit_tuple)
print(type(fruit_tuple))
fruits_list2 = list(fruit_tuple)
print(fruits_list2)
print(type(fruits_list2))   
```

```python
tuple2 = ('banana', 'kiwi', 'mango', 'apple', 'grape', 'mango')
list1 = list(tuple2)
print(list1)
list1.pop()
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```

```python
# sets {} | set() -> conversion from list to set

-> It is an inbuild Data structure which does not allow duplicates.
-> it is unordered
-> It is Mutable

union -> eliminates duplicates and orders result in ascending
intersection -> shows matched data between the data sets
minus -> shows unmatched data from 1st data set
```

```python
numbers = {1, 2, 3, 4, 5,5,4,3,2,1}
print(numbers)  
a= {1,2,3}
b = {3,4,5}
print("--------------")
print(a.union(b))
print(a.intersection(b))
print(a.difference(b))


print("--------------")

print(a|b)
print(a&b)
print(a-b)

print("--------------")
a.add(6)
print(a)
a.remove(2)
print(a)
print(len(a))
```

```python
# Dictionary {} | dict()  

-> It stores data as key-value pairs.
-> It is Ordered. 
-> It is Mutable.
-> Does not allow Duplicate Key.
```

```python
str = "vivek"
#stuid =[1,2]
#firstnames = ["vivek","srini"]
#lastnames = ["kumar","reddy"]

dict1 = {"studid": 100, "firstname": "vivek", "lastname": "kumar"}
print(dict1)
print(type(dict1))
```

```python
employee = {"emp_id": 101, "name": "Asha", "department": "IT", "salary": 65000}
print(employee)
```

```python
# accessing values from dictionary
print(employee["name"])
print(employee.get("department"))
```

```python
# adding and updating
# employee["salary"] = 70000
#print(employee["salary"])
#employee["location"] = "Bangalore"
print(employee)
employee.update({"salary": 75000, "location": "Hyderabad"})
print(employee)
```

```python
# remove 
print(employee.pop("salary"))
print(employee)
print(employee.popitem())
print(employee)
```

```python
# addition methods
print(employee)
print(employee.keys())
print(employee.values())
print(employee.items())
```
