```python
# while loop

counter = 0 
while counter <= 10:
    print(counter)
    counter += 1
    if counter %4 == 0:
        print("even number")
        break
    print("Counter value ",counter)
    
    
print("loop ends")
```

```python
# break - once it statisfies the condition it will break out of the loop
# continue

# while loop

counter = 0
while counter <= 10:
    counter += 1
    print("Counter value ",counter)    
    if counter %2 == 0:        
        print("even number:",counter)
        continue 
    print("Odd number:",counter)     
print("loop ends")
```

```python
menu  = {"idly":30,"vada":15,"coffee":15}

print(menu.get("idly"))
print(menu.get("coffee"))

#   idly = 30
#   vada = 15
#   coffeee = 15


for i in menu:
    # print(value)
    print(i," = ",menu.get(i))
    
```

```python
# q quit/break
menu  = {"idly":30,"vada":15,"coffee":15}
print(type(menu))
isQuit = True
list1=[]
total = 0

while isQuit:
    item = input("Enter the menu:")
    print(item)
    
    if item == "q":
        isQuit = False
        break
        
    list1.append(item)
    
print(list1)

for i in list1:
    print(i)
    total += menu.get(i)
print(total)
```
